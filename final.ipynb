# SPLITING DATASET

```python
import random
from pathlib import Path

SEED = 42
random.seed(SEED)

ROOT = Path.home() / "ML/PROJECT/FracAtlas"
FRACTURED = ROOT / "images/Fractured"
NORMAL = ROOT / "images/Non_fractured"
LABELS = ROOT / "Annotations/YOLO"

fractured = []
normal = []

for image in FRACTURED.iterdir():
    if image.suffix.lower() not in {".jpg", ".jpeg", ".png"}:
        continue

    label = LABELS / f"{image.stem}.txt"

    if not label.exists() or not label.read_text().strip():
        raise ValueError(f"Invalid fractured annotation: {image.name}")

    fractured.append((image, label))

for image in NORMAL.iterdir():
    if image.suffix.lower() not in {".jpg", ".jpeg", ".png"}:
        continue

    label = LABELS / f"{image.stem}.txt"

    if not label.exists():
        raise ValueError(f"Missing annotation: {image.name}")

    # Keep the two images with bounding boxes.
    normal.append((image, label))

random.shuffle(fractured)
random.shuffle(normal)

# Undersample non-fractured images.
normal = normal[:len(fractured)]

samples = (
    [(img, lbl, 1) for img, lbl in fractured] +
    [(img, lbl, 0) for img, lbl in normal]
)

random.shuffle(samples)

n = len(samples)
train = samples[:int(0.8 * n)]
val = samples[int(0.8 * n):int(0.9 * n)]
test = samples[int(0.9 * n):]

print("Train:", len(train))
print("Validation:", len(val))
print("Test:", len(test))
print("Total:", n)
```

# REMOVING CORRUPTED FILES

```python
import os
import shutil
import warnings
from pathlib import Path
import cv2
import pandas as pd
from PIL import Image, ImageFile
from tqdm import tqdm

# Enforce strict JPEG EOF checking (triggers OSError on truncated JPEGs)
ImageFile.LOAD_TRUNCATED_IMAGES = False

dataset_root = Path("./FracAtlas")
images_directory = dataset_root / "images"
dataset_csv_path = dataset_root / "dataset.csv"
quarantine_dir = Path("./FracAtlas_corrupted_backup")
quarantine_dir.mkdir(parents=True, exist_ok=True)

def is_image_corrupted(img_path: Path) -> tuple[bool, str]:
    """
    Checks for:
      1. Zero-byte or unreadable files
      2. Truncated JPEG streams ('Premature end of JPEG file')
      3. OpenCV decode failures
    """
    try:
        if not img_path.exists() or img_path.stat().st_size == 0:
            return True, "Empty or missing file"

        # Strict PIL full pixel decode (catches premature JPEG EOF)
        with Image.open(img_path) as img:
            img.verify()
        with Image.open(img_path) as img:
            img.load()

        # Verify OpenCV grayscale read matches expected dimensions
        cv_img = cv2.imread(str(img_path), cv2.IMREAD_GRAYSCALE)
        if cv_img is None or cv_img.size == 0:
            return True, "OpenCV returned None"

        return False, "OK"
    except Exception as e:
        return True, str(e)

# 1. Scan all images in ./FracAtlas/images
all_images = [
    p for p in images_directory.rglob("*")
    if p.suffix.lower() in {".jpg", ".jpeg", ".png"}
]
print(f"[*] Scanning {len(all_images)} images in {images_directory}...")

corrupted_records = []
for img_path in tqdm(all_images, desc="Verifying JPEGs"):
    corrupted, reason = is_image_corrupted(img_path)
    if corrupted:
        corrupted_records.append({
            "file_name": img_path.name,
            "stem": img_path.stem,
            "original_path": str(img_path),
            "reason": reason
        })
        # Move to backup quarantine folder (change to img_path.unlink() to permanently delete)
        dest = quarantine_dir / img_path.name
        shutil.move(str(img_path), str(dest))

print(f"\n[+] Found and quarantined {len(corrupted_records)} corrupted images -> {quarantine_dir}/")
for rec in corrupted_records:
    print(f"    - {rec['file_name']} ({rec['reason']})")

# 2. Clean ./FracAtlas/dataset.csv and any existing ./splits/*.csv
if corrupted_records:
    bad_names = {r["file_name"] for r in corrupted_records}
    bad_stems = {r["stem"] for r in corrupted_records}
    bad_tokens = bad_names | bad_stems

    # Backup & clean dataset.csv
    if dataset_csv_path.exists():
        df_meta = pd.read_csv(dataset_csv_path)
        shutil.copy2(dataset_csv_path, quarantine_dir / "dataset_backup.csv")
        before = len(df_meta)
        df_meta_clean = df_meta[~df_meta["image_id"].astype(str).isin(bad_tokens)].reset_index(drop=True)
        df_meta_clean.to_csv(dataset_csv_path, index=False)
        print(f"[+] Cleaned dataset.csv: {before} -> {len(df_meta_clean)} rows")

    # Clean existing split CSVs in ./splits/ if present
    splits_dir = Path("./splits")
    if splits_dir.exists():
        for csv_file in splits_dir.glob("*.csv"):
            df_split = pd.read_csv(csv_file)
            if "file_name" in df_split.columns:
                before_s = len(df_split)
                df_split = df_split[~df_split["file_name"].astype(str).isin(bad_tokens)].reset_index(drop=True)
                df_split.to_csv(csv_file, index=False)
                print(f"[+] Cleaned {csv_file.name}: {before_s} -> {len(df_split)} rows")
else:
    print("[+] No corrupted images detected.")
```

# VISUAL CHECK OF UNET++ (WITH RESNET ENCODER)

```python
import os
os.environ["OPENCV_LOG_LEVEL"] = "SILENT"

import json
import random
from pathlib import Path
import cv2
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import torch
import segmentation_models_pytorch as smp

# =====================================================================
# 1. Setup & Paths (Matches Your Local Workspace)
# =====================================================================
SEED = 42
TARGET_SHAPE = (512, 512)
random.seed(SEED)
np.random.seed(SEED)
torch.manual_seed(SEED)

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print(f"[+] Using device: {device}")

dataset_root = Path("./FracAtlas")
images_directory = dataset_root / "images"
weights_path = Path("./FracAtlas_UNetPP/unetpp_resnet34_best.pt")
test_split_path = Path("./splits/test_split_50_50.csv")
output_fig_path = Path("./FracAtlas_UNetPP/unetpp_30_fracture_comparison.png")

if not weights_path.exists():
    raise FileNotFoundError(f"Trained weights not found at {weights_path}")

# Locate COCO JSON
json_candidates = list((dataset_root / "Annotations").rglob("*COCO*.json"))
if not json_candidates:
    json_candidates = list((dataset_root / "Annotations").rglob("*.json"))
coco_json_path = json_candidates[0]

with open(coco_json_path, "r") as f:
    coco_payload = json.load(f)

filename_to_coco_img = {img["file_name"]: img for img in coco_payload["images"]}
img_id_to_annos = {}
for ann in coco_payload["annotations"]:
    img_id_to_annos.setdefault(ann["image_id"], []).append(ann)

# =====================================================================
# 2. Sample 30 Random Fractured Scans (Prefer Unseen Test Split)
# =====================================================================
if test_split_path.exists():
    df_source = pd.read_csv(test_split_path)
    df_frac = df_source[df_source["fractured"] == 1].reset_index(drop=True)
    split_label = "Hold-Out Test Split"
else:
    df_meta = pd.read_csv(dataset_root / "dataset.csv")
    rows = []
    for _, r in df_meta[df_meta["fractured"] == 1].iterrows():
        fname = str(r["image_id"])
        if not fname.lower().endswith((".jpg", ".jpeg", ".png")):
            fname += ".jpg"
        for sub in ["Fractured", "Non_fractured", ""]:
            cand = images_directory / sub / fname if sub else images_directory / fname
            if cand.exists():
                c_info = filename_to_coco_img.get(fname, {})
                rows.append({
                    "file_name": fname,
                    "path": str(cand),
                    "coco_id": c_info.get("id", -1),
                })
                break
    df_frac = pd.DataFrame(rows)
    split_label = "FracAtlas Fractured Cohort"

sample_30 = df_frac.sample(n=min(30, len(df_frac)), random_state=SEED).reset_index(drop=True)
print(f"[+] Sampled {len(sample_30)} fractured scans from {split_label}.")

# =====================================================================
# 3. Load Trained UNet++ (ResNet-34 Encoder)
# =====================================================================
model = smp.UnetPlusPlus(
    encoder_name="resnet34",
    encoder_weights=None,
    in_channels=3,
    classes=3,
).to(device)
model.load_state_dict(torch.load(weights_path, map_location=device))
model.eval()
print(f"[+] Loaded checkpoint: {weights_path}")

# =====================================================================
# 4. Preprocessing, Ground-Truth Rasterization & Dice Helper
# =====================================================================
clahe = cv2.createCLAHE(clipLimit=2.0, tileGridSize=(8, 8))

def prepare_scan_and_gt(img_path, coco_id):
    gray = cv2.imread(str(img_path), cv2.IMREAD_GRAYSCALE)
    if gray is None:
        raise FileNotFoundError(f"Cannot read {img_path}")
    orig_h, orig_w = gray.shape

    enhanced = clahe.apply(gray)
    resized = cv2.resize(enhanced, TARGET_SHAPE, interpolation=cv2.INTER_CUBIC)
    norm = (resized.astype(np.float32) - resized.min()) / (resized.max() - resized.min() + 1e-8)
    rgb = np.repeat(norm[:, :, np.newaxis], 3, axis=-1)

    # Ground-truth fracture mask (binary 0/1)
    gt_native = np.zeros((orig_h, orig_w), dtype=np.uint8)
    if coco_id != -1 and coco_id in img_id_to_annos:
        for ann in img_id_to_annos[coco_id]:
            seg = ann.get("segmentation", [])
            if isinstance(seg, list):
                for poly in seg:
                    pts = np.array(poly, dtype=np.int32).reshape((-1, 2))
                    cv2.fillPoly(gt_native, [pts], color=1)

    gt_resized = cv2.resize(gt_native, TARGET_SHAPE, interpolation=cv2.INTER_NEAREST)
    return rgb, gt_resized

def binary_dice(pred_bin, gt_bin, smooth=1e-5):
    inter = np.logical_and(pred_bin > 0, gt_bin > 0).sum()
    denom = (pred_bin > 0).sum() + (gt_bin > 0).sum()
    return (2.0 * inter + smooth) / (denom + smooth)

def overlay_mask(rgb_img, mask_bin, color=(1.0, 0.15, 0.15), alpha=0.55):
    out = rgb_img.copy()
    fg = mask_bin > 0
    for c in range(3):
        out[:, :, c] = np.where(fg, (1 - alpha) * out[:, :, c] + alpha * color[c], out[:, :, c])
    return np.clip(out, 0.0, 1.0)

# =====================================================================
# 5. Run Inference & Plot 30 x 3 Side-by-Side Comparison Grid
# =====================================================================
fig, axes = plt.subplots(nrows=len(sample_30), ncols=3, figsize=(13, 4.2 * len(sample_30)))
dice_scores = []

with torch.no_grad():
    for idx, row in sample_30.iterrows():
        rgb_img, gt_frac = prepare_scan_and_gt(row["path"], row["coco_id"])
        inp_tensor = torch.from_numpy(rgb_img).permute(2, 0, 1).unsqueeze(0).float().to(device)

        logits = model(inp_tensor)
        pred_classes = torch.argmax(logits, dim=1).squeeze(0).cpu().numpy()
        pred_frac = (pred_classes == 2).astype(np.uint8)

        dsc = binary_dice(pred_frac, gt_frac)
        dice_scores.append(dsc)

        # Col 1: Input Radiograph
        axes[idx, 0].imshow(rgb_img)
        axes[idx, 0].set_title(f"[{idx+1:02d}] Input: {row['file_name']}", fontsize=9, fontweight="bold")
        axes[idx, 0].axis("off")

        # Col 2: Ground-Truth Fracture Overlay (Red)
        gt_vis = overlay_mask(rgb_img, gt_frac, color=(1.0, 0.1, 0.1), alpha=0.60)
        axes[idx, 1].imshow(gt_vis)
        axes[idx, 1].set_title("Ground Truth Fracture (Red)", fontsize=9, color="#b30000")
        axes[idx, 1].axis("off")

        # Col 3: UNet++ Predicted Fracture Overlay (Green)
        pred_vis = overlay_mask(rgb_img, pred_frac, color=(0.1, 0.95, 0.2), alpha=0.60)
        axes[idx, 2].imshow(pred_vis)
        axes[idx, 2].set_title(f"UNet++ Pred (Green) | Dice: {dsc:.3f}", fontsize=9, color="#006600", fontweight="bold")
        axes[idx, 2].axis("off")

plt.suptitle(
    f"UNet++ (ResNet-34) Fracture Segmentation — 30 Random Unseen Scans (Mean Dice: {np.mean(dice_scores):.4f})",
    fontsize=13,
    fontweight="bold",
    y=1.001,
)
plt.tight_layout()
plt.savefig(output_fig_path, dpi=150, bbox_inches="tight")
print(f"[+] Saved 30-sample comparison image to: {output_fig_path}")
print(f"[+] Mean Fracture Dice across these 30 scans: {np.mean(dice_scores):.4f}")
```

# UNET++ (WITH RESNET ENCODER AND FOCAL TREVERSKY LOSS)

```python
import os
os.environ["OPENCV_LOG_LEVEL"] = "SILENT"

import json
import random
from pathlib import Path
import cv2
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import torch
import segmentation_models_pytorch as smp
from tqdm import tqdm

# =====================================================================
# 1. Setup & Paths (Matched to Your Local Workspace)
# =====================================================================
SEED = 42
TARGET_SHAPE = (512, 512)
BATCH_SIZE = 4

random.seed(SEED)
np.random.seed(SEED)
torch.manual_seed(SEED)
torch.cuda.manual_seed_all(SEED)

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print(f"[+] Running on Device: {device} ({torch.cuda.get_device_name(0) if torch.cuda.is_available() else 'CPU'})")

dataset_root = Path("./FracAtlas")
splits_dir = Path("./splits")
model_dir = Path("./FracAtlas_UNetPP_Focal_Treversky")
weights_path = model_dir / "Focal-Tversky + Weighted Cross-Entropy_unetpp.pt"
output_fig_path = model_dir / "focal_tversky_30_fracture_comparison.png"

if not weights_path.exists():
    raise FileNotFoundError(f"Checkpoint not found at: {weights_path}")

json_candidates = list((dataset_root / "Annotations").rglob("*COCO*.json")) or list((dataset_root / "Annotations").rglob("*.json"))
with open(json_candidates[0], "r") as f:
    coco_payload = json.load(f)

img_id_to_annos = {}
for ann in coco_payload["annotations"]:
    img_id_to_annos.setdefault(ann["image_id"], []).append(ann)

# =====================================================================
# 2. Load Trained UNet++ (ResNet-34, Focal-Tversky + WCE)
# =====================================================================
model = smp.UnetPlusPlus(
    encoder_name="resnet34",
    encoder_weights=None,
    in_channels=3,
    classes=3,
).to(device)

state_dict = torch.load(weights_path, map_location=device)
# Strip _orig_mod. prefix if saved from a torch.compile() wrapper
clean_state_dict = {k.replace("_orig_mod.", ""): v for k, v in state_dict.items()}
model.load_state_dict(clean_state_dict)
model.eval()
print(f"[+] Loaded Trained Weights: {weights_path}")

# =====================================================================
# 3. Deterministic Preprocessing & 3-Class Ground Truth Builder
# =====================================================================
clahe = cv2.createCLAHE(clipLimit=2.0, tileGridSize=(8, 8))

def prepare_sample(row):
    gray = cv2.imread(str(row["path"]), cv2.IMREAD_GRAYSCALE)
    if gray is None:
        raise FileNotFoundError(f"Cannot read image: {row['path']}")
    orig_h, orig_w = gray.shape

    enhanced = clahe.apply(gray)
    resized = cv2.resize(enhanced, TARGET_SHAPE, interpolation=cv2.INTER_CUBIC)
    norm = (resized.astype(np.float32) - resized.min()) / (resized.max() - resized.min() + 1e-8)
    rgb = np.repeat(norm[:, :, np.newaxis], 3, axis=-1)

    # Bone pseudo-label (Class 1)
    blurred = cv2.GaussianBlur(resized, (7, 7), 1.5)
    _, binary = cv2.threshold(blurred, 0, 255, cv2.THRESH_BINARY + cv2.THRESH_OTSU)
    kernel = cv2.getStructuringElement(cv2.MORPH_ELLIPSE, (9, 9))
    closed = cv2.morphologyEx(binary, cv2.MORPH_CLOSE, kernel, iterations=2)
    num_labels, labels, stats, _ = cv2.connectedComponentsWithStats(closed, connectivity=8)
    bone_mask = np.zeros_like(closed, dtype=np.uint8)
    for lbl in range(1, num_labels):
        if stats[lbl, cv2.CC_STAT_AREA] >= 450:
            bone_mask[labels == lbl] = 1

    # Ground-truth fracture polygons (Class 2)
    frac_native = np.zeros((orig_h, orig_w), dtype=np.uint8)
    coco_id = int(row["coco_id"])
    if coco_id != -1 and coco_id in img_id_to_annos:
        for ann in img_id_to_annos[coco_id]:
            seg = ann.get("segmentation", [])
            if isinstance(seg, list):
                for poly in seg:
                    pts = np.array(poly, dtype=np.int32).reshape((-1, 2))
                    cv2.fillPoly(frac_native, [pts], color=1)

    frac_resized = cv2.resize(frac_native, TARGET_SHAPE, interpolation=cv2.INTER_NEAREST)

    mask_3class = np.zeros(TARGET_SHAPE, dtype=np.int64)
    mask_3class[bone_mask > 0] = 1
    mask_3class[frac_resized > 0] = 2

    return rgb, mask_3class, frac_resized

# =====================================================================
# 4. Evaluate Train, Validation & Test Dice Scores
# =====================================================================
def evaluate_split_metrics(split_label, csv_path):
    df = pd.read_csv(csv_path).reset_index(drop=True)
    bone_dices, frac_dices, frac_ious, fp_healthy = [], [], [], []

    with torch.no_grad():
        for start in tqdm(range(0, len(df), BATCH_SIZE), desc=f"Evaluating {split_label}", leave=False):
            batch_df = df.iloc[start : start + BATCH_SIZE]
            imgs_list, masks_list, flags_list = [], [], []

            for _, row in batch_df.iterrows():
                rgb, m3, _ = prepare_sample(row)
                imgs_list.append(torch.from_numpy(rgb).permute(2, 0, 1).float())
                masks_list.append(torch.from_numpy(m3).long())
                flags_list.append(int(row["fractured"]))

            imgs_t = torch.stack(imgs_list).to(device)
            masks_t = torch.stack(masks_list).to(device)
            flags_t = torch.tensor(flags_list, device=device)

            with torch.amp.autocast("cuda", enabled=(device.type == "cuda")):
                logits = model(imgs_t)

            preds = torch.argmax(logits, dim=1)

            # Bone Dice (all images)
            p1, t1 = (preds == 1).float(), (masks_t == 1).float()
            bone_dices.append(((2.0 * (p1 * t1).sum() + 1e-5) / (p1.sum() + t1.sum() + 1e-5)).item())

            # Fracture Dice & IoU (fractured images only)
            pos_idx = (flags_t == 1)
            if pos_idx.sum() > 0:
                p2, t2 = (preds[pos_idx] == 2).float(), (masks_t[pos_idx] == 2).float()
                inter = (p2 * t2).sum()
                union = p2.sum() + t2.sum() - inter
                frac_dices.append(((2.0 * inter + 1e-5) / (p2.sum() + t2.sum() + 1e-5)).item())
                frac_ious.append(((inter + 1e-5) / (union + 1e-5)).item())

            # Healthy False-Positive Pixels
            neg_idx = (flags_t == 0)
            if neg_idx.sum() > 0:
                fp_neg = (preds[neg_idx] == 2).float().sum(dim=(1, 2)).mean().item()
                fp_healthy.append(fp_neg)

    return {
        "n_scans": len(df),
        "bone_dice": np.mean(bone_dices),
        "frac_dice": np.mean(frac_dices),
        "frac_iou": np.mean(frac_ious),
        "fp_px": np.mean(fp_healthy) if fp_healthy else 0.0,
    }

split_files = [
    ("Train (70%)", splits_dir / "train_split_50_50.csv"),
    ("Val   (15%)", splits_dir / "val_split_50_50.csv"),
    ("Test  (15%)", splits_dir / "test_split_50_50.csv"),
]

summary = {name: evaluate_split_metrics(name, path) for name, path in split_files}

print("\n============= UNet++ (Focal-Tversky + WCE) SPLIT EVALUATION =============")
print(f"{'Split':<13} | {'Scans':<6} | {'Bone Dice':<10} | {'Frac Dice':<10} | {'Frac IoU':<9} | {'Healthy FP (px)'}")
print("-" * 73)
for name, m in summary.items():
    print(f"{name:<13} | {m['n_scans']:<6} | {m['bone_dice']:<10.4f} | {m['frac_dice']:<10.4f} | {m['frac_iou']:<9.4f} | {m['fp_px']:>8.2f}")
print("=" * 73)

# =====================================================================
# 5. Sample 30 Random Fractured Scans & Plot Ground Truth vs. Output
# =====================================================================
test_df = pd.read_csv(splits_dir / "test_split_50_50.csv")
frac_test_df = test_df[test_df["fractured"] == 1].reset_index(drop=True)
sample_30 = frac_test_df.sample(n=min(30, len(frac_test_df)), random_state=SEED).reset_index(drop=True)

def overlay_mask(rgb_img, mask_bin, color, alpha=0.58):
    out = rgb_img.copy()
    fg = mask_bin > 0
    for c in range(3):
        out[:, :, c] = np.where(fg, (1.0 - alpha) * out[:, :, c] + alpha * color[c], out[:, :, c])
    return np.clip(out, 0.0, 1.0)

def single_scan_dice(pred_bin, gt_bin, smooth=1e-5):
    inter = np.logical_and(pred_bin > 0, gt_bin > 0).sum()
    denom = (pred_bin > 0).sum() + (gt_bin > 0).sum()
    return (2.0 * inter + smooth) / (denom + smooth)

print(f"\n[*] Generating side-by-side comparison for {len(sample_30)} random fractured test scans...")
fig, axes = plt.subplots(nrows=len(sample_30), ncols=3, figsize=(13, 4.2 * len(sample_30)))
sample_dices = []

with torch.no_grad():
    for idx, row in tqdm(sample_30.iterrows(), total=len(sample_30), desc="Rendering 30 Comparisons"):
        rgb, _, gt_frac = prepare_sample(row)
        inp = torch.from_numpy(rgb).permute(2, 0, 1).unsqueeze(0).float().to(device)

        with torch.amp.autocast("cuda", enabled=(device.type == "cuda")):
            logits = model(inp)

        pred_classes = torch.argmax(logits, dim=1).squeeze(0).cpu().numpy()
        pred_frac = (pred_classes == 2).astype(np.uint8)

        dsc = single_scan_dice(pred_frac, gt_frac)
        sample_dices.append(dsc)

        # Col 1: Input X-Ray
        axes[idx, 0].imshow(rgb)
        axes[idx, 0].set_title(f"[{idx+1:02d}] Input: {row['file_name']}", fontsize=9, fontweight="bold")
        axes[idx, 0].axis("off")

        # Col 2: Ground-Truth Fracture Mask (Red Overlay)
        gt_overlay = overlay_mask(rgb, gt_frac, color=(1.0, 0.1, 0.1), alpha=0.60)
        axes[idx, 1].imshow(gt_overlay)
        axes[idx, 1].set_title("Ground Truth Fracture (Red)", fontsize=9, color="#b30000")
        axes[idx, 1].axis("off")

        # Col 3: Model Predicted Fracture Mask (Green Overlay)
        pred_overlay = overlay_mask(rgb, pred_frac, color=(0.1, 0.95, 0.2), alpha=0.60)
        axes[idx, 2].imshow(pred_overlay)
        axes[idx, 2].set_title(f"Focal-Tversky UNet++ Output | Dice: {dsc:.3f}", fontsize=9, color="#006600", fontweight="bold")
        axes[idx, 2].axis("off")

plt.suptitle(
    f"Focal-Tversky + WCE UNet++ (ResNet-34) — 30 Random Unseen Fractured Scans (Mean Dice: {np.mean(sample_dices):.4f})",
    fontsize=13,
    fontweight="bold",
    y=1.001,
)
plt.tight_layout()
plt.savefig(output_fig_path, dpi=150, bbox_inches="tight")
print(f"[+] Saved 30-image comparison grid to: {output_fig_path}")
print(f"[+] Mean Fracture Dice on these 30 scans: {np.mean(sample_dices):.4f}")
```

```python
import os
os.environ["OPENCV_LOG_LEVEL"] = "SILENT"
os.environ["PYTORCH_CUDA_ALLOC_CONF"] = "expandable_segments:True"

import json
import math
import time
import random
import urllib.request
from pathlib import Path
from concurrent.futures import ThreadPoolExecutor

import cv2
import numpy as np
import pandas as pd
from tqdm import tqdm

import torch
import torch.nn as nn
import torch.nn.functional as F

try:
    from segment_anything import sam_model_registry
except ImportError as e:
    raise ImportError(
        "segment_anything is not installed. Run: pip install git+https://github.com/facebookresearch/segment-anything.git"
    ) from e

# =====================================================================
# 1. Global Constants & RTX 4050 Setup (SEED = 42, 512 x 512, Binary)
# =====================================================================
SEED = 42
TARGET_SHAPE = (512, 512)
BATCH_SIZE = 4
ACCUM_STEPS = 2          # Effective batch size = 8
EPOCHS = 25
LR_LORA = 1.5e-4
LR_DECODER = 3.0e-4
LORA_RANK = 8
LORA_ALPHA = 16
LORA_DROPOUT = 0.05

def seed_everything(seed=42):
    random.seed(seed)
    os.environ["PYTHONHASHSEED"] = str(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)
    torch.cuda.manual_seed_all(seed)
    torch.backends.cudnn.benchmark = True
    torch.set_float32_matmul_precision("high")

seed_everything(SEED)
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
use_bf16 = torch.cuda.is_available() and torch.cuda.is_bf16_supported()
amp_dtype = torch.bfloat16 if use_bf16 else torch.float16
print(f"[+] Device: {device} ({torch.cuda.get_device_name(0) if torch.cuda.is_available() else 'CPU'}) | 512x512 Binary | AMP: {amp_dtype}")

# =====================================================================
# 2. Paths & Non-Destructive Checkpoint Directory
# =====================================================================
dataset_root = Path("./FracAtlas")
splits_dir = Path("./splits")
save_dir = Path("./FracAtlas_MedSAM_LoRA")
save_dir.mkdir(parents=True, exist_ok=True)
best_weights_path = save_dir / "medsam_lora_r8_binary_best.pt"

train_csv = splits_dir / "train_split_50_50.csv"
val_csv   = splits_dir / "val_split_50_50.csv"
test_csv  = splits_dir / "test_split_50_50.csv"

if not (train_csv.exists() and val_csv.exists() and test_csv.exists()):
    raise FileNotFoundError("Saved 50/50 split CSVs not found in ./splits/.")

json_candidates = list((dataset_root / "Annotations").rglob("*COCO*.json")) or list((dataset_root / "Annotations").rglob("*.json"))
with open(json_candidates[0], "r") as f:
    coco_payload = json.load(f)

img_id_to_annos = {}
for ann in coco_payload["annotations"]:
    img_id_to_annos.setdefault(ann["image_id"], []).append(ann)

# =====================================================================
# 3. Parallel Preprocessing (2-Class: 0 = Background, 1 = Fracture)
# =====================================================================
def preprocess_split(csv_path, split_name):
    df = pd.read_csv(csv_path).reset_index(drop=True)
    H, W = TARGET_SHAPE

    def _worker(row_dict):
        clahe = cv2.createCLAHE(clipLimit=2.0, tileGridSize=(8, 8))
        gray = cv2.imread(str(row_dict["path"]), cv2.IMREAD_GRAYSCALE)
        if gray is None:
            raise FileNotFoundError(f"Cannot read image: {row_dict['path']}")
        orig_h, orig_w = gray.shape

        enhanced = clahe.apply(gray)
        resized = cv2.resize(enhanced, (W, H), interpolation=cv2.INTER_CUBIC)
        r_min, r_max = float(resized.min()), float(resized.max())
        norm_u8 = ((resized.astype(np.float32) - r_min) / (r_max - r_min + 1e-8) * 255.0).astype(np.uint8)

        # Ground-truth binary fracture mask (0: Background, 1: Fracture)
        frac_native = np.zeros((orig_h, orig_w), dtype=np.uint8)
        coco_id = int(row_dict["coco_id"])
        if coco_id != -1 and coco_id in img_id_to_annos:
            for ann in img_id_to_annos[coco_id]:
                seg = ann.get("segmentation", [])
                if isinstance(seg, list):
                    for poly in seg:
                        pts = np.array(poly, dtype=np.int32).reshape((-1, 2))
                        cv2.fillPoly(frac_native, [pts], color=1)

        frac_mask = cv2.resize(frac_native, (W, H), interpolation=cv2.INTER_NEAREST)

        # Prompt box [xmin, ymin, xmax, ymax] in 512x512 space
        ys, xs = np.where(frac_mask > 0)
        if len(xs) > 0 and len(ys) > 0:
            bbox = np.array([
                max(0, int(xs.min()) - 5),
                max(0, int(ys.min()) - 5),
                min(W - 1, int(xs.max()) + 5),
                min(H - 1, int(ys.max()) + 5),
            ], dtype=np.float32)
        else:
            # Healthy scan: central anatomical region prompt with all-zero target
            bbox = np.array([96.0, 96.0, W - 96.0, H - 96.0], dtype=np.float32)

        return norm_u8, frac_mask, bbox, int(row_dict["fractured"])

    records = df.to_dict(orient="records")
    with ThreadPoolExecutor(max_workers=min(8, os.cpu_count() or 4)) as ex:
        results = list(tqdm(ex.map(_worker, records), total=len(records), desc=f"Caching {split_name} -> RAM"))

    imgs_t  = torch.from_numpy(np.stack([r[0] for r in results])).unsqueeze(1)  # (N, 1, 512, 512) uint8
    masks_t = torch.from_numpy(np.stack([r[1] for r in results])).unsqueeze(1)  # (N, 1, 512, 512) uint8
    boxes_t = torch.from_numpy(np.stack([r[2] for r in results]))               # (N, 4) float32
    flags_t = torch.tensor([r[3] for r in results], dtype=torch.long)

    if device.type == "cuda":
        imgs_t = imgs_t.pin_memory()
        masks_t = masks_t.pin_memory()
    return imgs_t, masks_t, boxes_t, flags_t

train_imgs, train_masks, train_boxes, train_flags = preprocess_split(train_csv, "Train (70%)")
val_imgs,   val_masks,   val_boxes,   val_flags   = preprocess_split(val_csv,   "Val   (15%)")
test_imgs,  test_masks,  test_boxes,  test_flags  = preprocess_split(test_csv,  "Test  (15%)")

# =====================================================================
# 4. Binary MedSAM ViT-B + 512x512 Pos-Embed Interpolation + LoRA (r=8)
# =====================================================================
ckpt_path = Path("./medsam_vit_b.pth")
if not ckpt_path.exists():
    print("[*] Downloading medsam_vit_b.pth (~375 MB)...")
    urllib.request.urlretrieve(
        "https://zenodo.org/records/10689643/files/medsam_vit_b.pth?download=1", str(ckpt_path)
    )

def resize_medsam_to_512(sam_model, target_size=512):
    new_grid = target_size // 16  # 32
    sam_model.image_encoder.img_size = target_size
    sam_model.prompt_encoder.input_image_size = (target_size, target_size)
    sam_model.prompt_encoder.image_embedding_size = (new_grid, new_grid)

    pos_embed = sam_model.image_encoder.pos_embed.permute(0, 3, 1, 2)
    pos_embed_new = F.interpolate(pos_embed, size=(new_grid, new_grid), mode="bicubic", align_corners=False)
    sam_model.image_encoder.pos_embed = nn.Parameter(pos_embed_new.permute(0, 2, 3, 1))

    new_rel_len = 2 * new_grid - 1  # 63
    for blk in sam_model.image_encoder.blocks:
        if hasattr(blk.attn, "rel_pos_h") and blk.attn.rel_pos_h is not None:
            if blk.attn.rel_pos_h.shape[0] != new_rel_len:
                rh = blk.attn.rel_pos_h.permute(1, 0).unsqueeze(0)
                rh_new = F.interpolate(rh, size=new_rel_len, mode="linear", align_corners=False)
                blk.attn.rel_pos_h = nn.Parameter(rh_new.squeeze(0).permute(1, 0))

                rw = blk.attn.rel_pos_w.permute(1, 0).unsqueeze(0)
                rw_new = F.interpolate(rw, size=new_rel_len, mode="linear", align_corners=False)
                blk.attn.rel_pos_w = nn.Parameter(rw_new.squeeze(0).permute(1, 0))

class LoRAQKVLinear(nn.Module):
    def __init__(self, qkv_linear: nn.Linear, r: int = 8, alpha: int = 16, dropout: float = 0.05):
        super().__init__()
        self.qkv = qkv_linear
        for p in self.qkv.parameters():
            p.requires_grad = False

        in_dim = qkv_linear.in_features
        out_dim = qkv_linear.out_features
        self.scaling = alpha / float(r)
        self.drop = nn.Dropout(dropout) if dropout > 0 else nn.Identity()

        self.lora_A = nn.Parameter(torch.zeros((r, in_dim)))
        self.lora_B = nn.Parameter(torch.zeros((out_dim, r)))
        nn.init.kaiming_uniform_(self.lora_A, a=math.sqrt(5))
        nn.init.zeros_(self.lora_B)

    def forward(self, x):
        base_out = self.qkv(x)
        lora_out = F.linear(F.linear(self.drop(x), self.lora_A), self.lora_B) * self.scaling
        return base_out + lora_out

class MedSAMLoRABinary(nn.Module):
    """
    MedSAM ViT-B with LoRA (r=8) on Image Encoder Attention +
    Native Single-Channel Binary Mask Decoder (0: Background, 1: Fracture).
    """
    def __init__(self, checkpoint_path, rank=8, alpha=16, dropout=0.05):
        super().__init__()
        self.sam = sam_model_registry["vit_b"](checkpoint=str(checkpoint_path))
        resize_medsam_to_512(self.sam, target_size=TARGET_SHAPE[0])

        for p in self.sam.image_encoder.parameters():
            p.requires_grad = False
        for p in self.sam.prompt_encoder.parameters():
            p.requires_grad = False

        for blk in self.sam.image_encoder.blocks:
            blk.attn.qkv = LoRAQKVLinear(blk.attn.qkv, r=rank, alpha=alpha, dropout=dropout)

        for p in self.sam.mask_decoder.parameters():
            p.requires_grad = True

    def forward(self, images_512, boxes_512):
        image_embeddings = self.sam.image_encoder(images_512)  # (B, 256, 32, 32)
        dense_pe = self.sam.prompt_encoder.get_dense_pe()      # (1, 256, 32, 32)

        sparse_emb, dense_emb = self.sam.prompt_encoder(
            points=None, boxes=boxes_512.unsqueeze(1), masks=None
        )

        # Batched MaskDecoder pass (1 box prompt per image in batch B)
        dec = self.sam.mask_decoder
        output_tokens = torch.cat([dec.iou_token.weight, dec.mask_tokens.weight], dim=0)  # (5, 256)
        output_tokens = output_tokens.unsqueeze(0).expand(sparse_emb.size(0), -1, -1)
        tokens = torch.cat((output_tokens, sparse_emb), dim=1)

        src = image_embeddings + dense_emb
        pos_src = torch.repeat_interleave(dense_pe, tokens.shape[0], dim=0)
        b, c, h, w = src.shape

        hs, src = dec.transformer(src, pos_src, tokens)
        mask_tokens_out = hs[:, 1 : 2, :]  # First mask token (B, 1, 256)

        src = src.transpose(1, 2).view(b, c, h, w)
        upscaled_embedding = dec.output_upscaling(src)  # (B, 32, 128, 128)
        hyper_in = dec.output_hypernetworks_mlps[0](mask_tokens_out[:, 0, :]).unsqueeze(1)  # (B, 1, 32)

        b, c_up, h_up, w_up = upscaled_embedding.shape
        low_res_logits = (hyper_in @ upscaled_embedding.view(b, c_up, h_up * w_up)).view(b, 1, h_up, w_up)

        return F.interpolate(low_res_logits, size=TARGET_SHAPE, mode="bilinear", align_corners=False)

model = MedSAMLoRABinary(ckpt_path, rank=LORA_RANK, alpha=LORA_ALPHA, dropout=LORA_DROPOUT).to(device)

total_params = sum(p.numel() for p in model.parameters())
trainable_params = sum(p.numel() for p in model.parameters() if p.requires_grad)
print(f"[+] Binary MedSAM + LoRA (r={LORA_RANK}) Initialized:")
print(f"    Total Parameters:     {total_params:,}")
print(f"    Trainable Parameters: {trainable_params:,} ({trainable_params / total_params * 100:.2f}%)")

# =====================================================================
# 5. Binary BCE + Focal-Tversky Loss (alpha=0.7, beta=0.3, gamma=1.33)
# =====================================================================
criterion_bce = nn.BCEWithLogitsLoss()

def binary_focal_tversky_loss(logits, targets, alpha=0.7, beta=0.3, gamma=1.33, smooth=1e-5):
    probs = torch.sigmoid(logits.float())
    targets_f = targets.float()

    tp = torch.sum(probs * targets_f)
    fn = torch.sum((1.0 - probs) * targets_f)
    fp = torch.sum(probs * (1.0 - targets_f))

    tversky = (tp + smooth) / (tp + alpha * fn + beta * fp + smooth)
    return torch.pow(1.0 - tversky, gamma)

def compute_binary_dice(logits, targets, smooth=1e-5):
    preds = (torch.sigmoid(logits) > 0.5).float()
    targets_f = targets.float()
    inter = (preds * targets_f).sum()
    denom = preds.sum() + targets_f.sum()
    return ((2.0 * inter + smooth) / (denom + smooth)).item()

lora_params = [p for n, p in model.named_parameters() if p.requires_grad and "lora_" in n]
decoder_params = [p for n, p in model.named_parameters() if p.requires_grad and "lora_" not in n]

optimizer = torch.optim.AdamW(
    [
        {"params": lora_params, "lr": LR_LORA},
        {"params": decoder_params, "lr": LR_DECODER},
    ],
    weight_decay=1e-2,
    fused=(device.type == "cuda"),
)
scheduler = torch.optim.lr_scheduler.CosineAnnealingLR(optimizer, T_max=EPOCHS, eta_min=1e-6)
scaler = torch.amp.GradScaler("cuda", enabled=(device.type == "cuda" and not use_bf16))

# =====================================================================
# 6. Fine-Tuning Loop with Jittered Box Prompts & Progress Bars
# =====================================================================
best_val_frac_dice = 0.0
n_train = train_imgs.size(0)
n_val = val_imgs.size(0)

print(f"\n[*] Fine-Tuning Binary MedSAM + LoRA (r={LORA_RANK}) -> {best_weights_path}\n")
t_start = time.time()

for epoch in range(1, EPOCHS + 1):
    ep_t0 = time.time()
    model.train()
    running_loss = torch.zeros(1, device=device)
    num_batches = 0
    optimizer.zero_grad(set_to_none=True)

    perm = torch.randperm(n_train)
    total_steps = (n_train + BATCH_SIZE - 1) // BATCH_SIZE

    train_pbar = tqdm(
        range(0, n_train, BATCH_SIZE),
        total=total_steps,
        desc=f"Epoch {epoch:02d}/{EPOCHS:02d} [Train]",
        leave=False,
        mininterval=0.5,
    )

    for step, start in enumerate(train_pbar):
        idx = perm[start : start + BATCH_SIZE]
        imgs = train_imgs[idx].to(device, non_blocking=True).to(dtype=amp_dtype).div_(255.0)
        masks = train_masks[idx].to(device, non_blocking=True).float()
        boxes = train_boxes[idx].clone().to(device, non_blocking=True)

        # Random Horizontal Flip
        if random.random() > 0.5:
            imgs = torch.flip(imgs, dims=[3])
            masks = torch.flip(masks, dims=[3])
            old_x1 = boxes[:, 0].clone()
            old_x2 = boxes[:, 2].clone()
            boxes[:, 0] = (TARGET_SHAPE[1] - 1) - old_x2
            boxes[:, 2] = (TARGET_SHAPE[1] - 1) - old_x1

        # Random Box Jitter (+/- 18 px)
        jitter = torch.empty_like(boxes).uniform_(-18.0, 18.0)
        boxes[:, 0] = torch.clamp(boxes[:, 0] + jitter[:, 0], 0.0, float(TARGET_SHAPE[1] - 20))
        boxes[:, 1] = torch.clamp(boxes[:, 1] + jitter[:, 1], 0.0, float(TARGET_SHAPE[0] - 20))
        boxes[:, 2] = torch.clamp(boxes[:, 2] + jitter[:, 2], min=boxes[:, 0] + 12.0, max=torch.tensor(float(TARGET_SHAPE[1] - 1), device=device))
        boxes[:, 3] = torch.clamp(boxes[:, 3] + jitter[:, 3], min=boxes[:, 1] + 12.0, max=torch.tensor(float(TARGET_SHAPE[0] - 1), device=device))

        imgs = imgs.repeat(1, 3, 1, 1)

        with torch.amp.autocast("cuda", dtype=amp_dtype, enabled=(device.type == "cuda")):
            logits = model(imgs, boxes)
            loss = (criterion_bce(logits, masks) + binary_focal_tversky_loss(logits, masks)) / ACCUM_STEPS

        if scaler.is_enabled():
            scaler.scale(loss).backward()
            if (step + 1) % ACCUM_STEPS == 0 or (step + 1) == total_steps:
                scaler.step(optimizer)
                scaler.update()
                optimizer.zero_grad(set_to_none=True)
        else:
            loss.backward()
            if (step + 1) % ACCUM_STEPS == 0 or (step + 1) == total_steps:
                optimizer.step()
                optimizer.zero_grad(set_to_none=True)

        running_loss += loss.detach() * ACCUM_STEPS
        num_batches += 1

    scheduler.step()
    avg_train_loss = (running_loss / num_batches).item()

    # Validation Pass
    model.eval()
    val_loss_t = torch.zeros(1, device=device)
    frac_dices = []
    val_batches = 0

    with torch.no_grad():
        for start in range(0, n_val, BATCH_SIZE):
            imgs = val_imgs[start : start + BATCH_SIZE].to(device, non_blocking=True).to(dtype=amp_dtype).div_(255.0)
            imgs = imgs.repeat(1, 3, 1, 1)
            masks = val_masks[start : start + BATCH_SIZE].to(device, non_blocking=True).float()
            boxes = val_boxes[start : start + BATCH_SIZE].to(device, non_blocking=True)
            flags = val_flags[start : start + BATCH_SIZE]

            with torch.amp.autocast("cuda", dtype=amp_dtype, enabled=(device.type == "cuda")):
                logits = model(imgs, boxes)
                loss = criterion_bce(logits, masks) + binary_focal_tversky_loss(logits, masks)

            val_loss_t += loss.detach()
            val_batches += 1
            if flags.sum() > 0:
                f_idx = (flags == 1).to(device)
                frac_dices.append(compute_binary_dice(logits[f_idx], masks[f_idx]))

    avg_val_loss = (val_loss_t / val_batches).item()
    mean_frac_dice = np.mean(frac_dices) if frac_dices else 0.0
    ep_sec = time.time() - ep_t0

    print(f"Epoch {epoch:02d}/{EPOCHS:02d} ({ep_sec:.1f}s) | Train Loss: {avg_train_loss:.4f} | "
          f"Val Loss: {avg_val_loss:.4f} | Val Frac Dice: {mean_frac_dice:.4f}")

    if mean_frac_dice > best_val_frac_dice:
        best_val_frac_dice = mean_frac_dice
        torch.save(model.state_dict(), best_weights_path)
        print(f"   --> Saved Best Binary MedSAM+LoRA Model (Val Fracture Dice: {best_val_frac_dice:.4f})")

print(f"\n[+] Finished 25 epochs in {(time.time() - t_start)/60:.2f} minutes!")

# =====================================================================
# 7. Hold-Out Test Split Evaluation (Box-Prompted & Full-Image Box)
# =====================================================================
model.load_state_dict(torch.load(best_weights_path, map_location=device))
model.eval()

test_frac_dices_box, test_frac_dices_full, fp_pixels_neg = [], [], []
n_test = test_imgs.size(0)
full_box_template = torch.tensor([[0.0, 0.0, 511.0, 511.0]], device=device)

with torch.no_grad():
    for start in tqdm(range(0, n_test, BATCH_SIZE), desc="Testing Binary MedSAM+LoRA"):
        imgs = test_imgs[start : start + BATCH_SIZE].to(device, non_blocking=True).to(dtype=amp_dtype).div_(255.0)
        imgs = imgs.repeat(1, 3, 1, 1)
        masks = test_masks[start : start + BATCH_SIZE].to(device, non_blocking=True).float()
        boxes = test_boxes[start : start + BATCH_SIZE].to(device, non_blocking=True)
        flags = test_flags[start : start + BATCH_SIZE]

        with torch.amp.autocast("cuda", dtype=amp_dtype, enabled=(device.type == "cuda")):
            logits_box = model(imgs, boxes)
            logits_full = model(imgs, full_box_template.expand(imgs.size(0), -1))

        pos_mask = (flags == 1).to(device)
        if pos_mask.sum() > 0:
            test_frac_dices_box.append(compute_binary_dice(logits_box[pos_mask], masks[pos_mask]))
            test_frac_dices_full.append(compute_binary_dice(logits_full[pos_mask], masks[pos_mask]))

        neg_mask = (flags == 0).to(device)
        if neg_mask.sum() > 0:
            preds_neg = (torch.sigmoid(logits_full[neg_mask]) > 0.5).float()
            fp_pixels_neg.append(preds_neg.sum(dim=(1, 2, 3)).mean().item())

print("\n==== FINAL TEST RESULTS: Binary MedSAM + LoRA (r=8) ====")
print(f"Saved Checkpoint:                                {best_weights_path}")
print(f"Trainable Parameter Share:                       {trainable_params / total_params * 100:.2f}%")
print(f"Test Fracture Dice (Box-Prompted):               {np.mean(test_frac_dices_box):.4f}")
print(f"Test Fracture Dice (Autonomous Full-Image Box):  {np.mean(test_frac_dices_full):.4f}")
print(f"Avg False-Positive Pixels (Healthy Scans):       {np.mean(fp_pixels_neg):.2f} px / image")
print("========================================================")
```

# EVALUATING (UNET++ & MedSAM CASCADE) WITH  (MASK FUSION), (TTA), (VARIOUS PADDING OPTIONS)

```python
import os
os.environ["OPENCV_LOG_LEVEL"] = "SILENT"
os.environ["PYTORCH_CUDA_ALLOC_CONF"] = "expandable_segments:True"

import json
import time
import random
from pathlib import Path
from concurrent.futures import ThreadPoolExecutor

import cv2
import numpy as np
import pandas as pd
import matplotlib
matplotlib.use("Agg")
import matplotlib.pyplot as plt
from tqdm import tqdm

import torch
import torch.nn as nn
import torch.nn.functional as F
import segmentation_models_pytorch as smp
from segment_anything import sam_model_registry

# =====================================================================
# 1. Workspace Paths & Fusion Hyperparameters (SEED = 42, 512 x 512)
# =====================================================================
SEED = 42
TARGET_SHAPE = (512, 512)
BOX_PADDING = 9
MIN_BLOB_AREA = 12
PROB_THRESH = 0.50
ALPHAS = [1.00, 0.75, 0.70, 0.65, 0.60]   # 1.00 = Pure MedSAM-LoRA (Baseline)

random.seed(SEED)
np.random.seed(SEED)
torch.manual_seed(SEED)
torch.cuda.manual_seed_all(SEED)

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
use_bf16 = torch.cuda.is_available() and torch.cuda.is_bf16_supported()
amp_dtype = torch.bfloat16 if use_bf16 else torch.float16
print(f"[+] Device: {device} ({torch.cuda.get_device_name(0) if torch.cuda.is_available() else 'CPU'}) | AMP: {amp_dtype}")

DATASET_ROOT = Path("./FracAtlas")
TEST_CSV = Path("./splits/test_split_50_50.csv")

UNETPP_DIR = Path("./FracAtlas_UNetPP_Focal_Treversky")
UNETPP_CKPT = UNETPP_DIR / "Focal-Tversky + Weighted Cross-Entropy_unetpp.pt"
if not UNETPP_CKPT.exists():
    cands = sorted(UNETPP_DIR.glob("*.pt")) + sorted(UNETPP_DIR.glob("*.pth"))
    if not cands:
        raise FileNotFoundError(f"No U-Net++ checkpoint found in {UNETPP_DIR}")
    UNETPP_CKPT = cands[0]

MEDSAM_BASE_CKPT = Path("./medsam_vit_b.pth")
MEDSAM_LORA_DIR = Path("./FracAtlas_MedSAM_LoRA")
MEDSAM_LORA_CKPT = MEDSAM_LORA_DIR / "medsam_lora_r8_binary_best.pt"

OUT_CSV_PATH = MEDSAM_LORA_DIR / "cascade_prob_fusion_benchmark.csv"
OUT_FIG_PATH = MEDSAM_LORA_DIR / "cascade_prob_fusion_20.png"

if not MEDSAM_LORA_CKPT.exists():
    raise FileNotFoundError(f"MedSAM+LoRA checkpoint not found at: {MEDSAM_LORA_CKPT}")
if not TEST_CSV.exists():
    raise FileNotFoundError(f"Hold-out test split CSV not found at: {TEST_CSV}")

def extract_clean_state_dict(ckpt_path):
    payload = torch.load(ckpt_path, map_location="cpu")
    state = payload["model_state_dict"] if (isinstance(payload, dict) and "model_state_dict" in payload) else payload
    return {k.replace("_orig_mod.", ""): v for k, v in state.items()}

# =====================================================================
# 2. Load Stage 1 (U-Net++ ResNet-34) & Stage 2 (MedSAM ViT-B + LoRA)
# =====================================================================
unetpp = smp.UnetPlusPlus(
    encoder_name="resnet34", encoder_weights=None, in_channels=3, classes=3
)
unetpp.load_state_dict(extract_clean_state_dict(UNETPP_CKPT))
unetpp = unetpp.float().to(device)
unetpp.eval()
print(f"[+] Stage 1 Loaded: {UNETPP_CKPT}")

def resize_medsam_to_512(sam_model, target_size=512):
    new_grid = target_size // 16
    sam_model.image_encoder.img_size = target_size
    sam_model.prompt_encoder.input_image_size = (target_size, target_size)
    sam_model.prompt_encoder.image_embedding_size = (new_grid, new_grid)

    pos_embed = sam_model.image_encoder.pos_embed.permute(0, 3, 1, 2)
    pos_embed_new = F.interpolate(pos_embed, size=(new_grid, new_grid), mode="bicubic", align_corners=False)
    sam_model.image_encoder.pos_embed = nn.Parameter(pos_embed_new.permute(0, 2, 3, 1))

    new_rel_len = 2 * new_grid - 1
    for blk in sam_model.image_encoder.blocks:
        if hasattr(blk.attn, "rel_pos_h") and blk.attn.rel_pos_h is not None:
            if blk.attn.rel_pos_h.shape[0] != new_rel_len:
                rh = blk.attn.rel_pos_h.permute(1, 0).unsqueeze(0)
                blk.attn.rel_pos_h = nn.Parameter(
                    F.interpolate(rh, size=new_rel_len, mode="linear", align_corners=False).squeeze(0).permute(1, 0)
                )
                rw = blk.attn.rel_pos_w.permute(1, 0).unsqueeze(0)
                blk.attn.rel_pos_w = nn.Parameter(
                    F.interpolate(rw, size=new_rel_len, mode="linear", align_corners=False).squeeze(0).permute(1, 0)
                )

class LoRAQKVLinear(nn.Module):
    def __init__(self, qkv_linear: nn.Linear, r: int = 8, alpha: int = 16):
        super().__init__()
        self.qkv = qkv_linear
        self.scaling = alpha / float(r)
        self.lora_A = nn.Parameter(torch.zeros((r, qkv_linear.in_features)))
        self.lora_B = nn.Parameter(torch.zeros((qkv_linear.out_features, r)))

    def forward(self, x):
        return self.qkv(x) + F.linear(F.linear(x, self.lora_A), self.lora_B) * self.scaling

class MedSAMLoRABinary(nn.Module):
    def __init__(self, checkpoint_path, rank=8, alpha=16):
        super().__init__()
        self.sam = sam_model_registry["vit_b"](checkpoint=str(checkpoint_path))
        resize_medsam_to_512(self.sam, target_size=TARGET_SHAPE[0])
        for blk in self.sam.image_encoder.blocks:
            blk.attn.qkv = LoRAQKVLinear(blk.attn.qkv, r=rank, alpha=alpha)

    @torch.inference_mode()
    def decode_boxes_prob_map(self, image_embeddings, boxes_list):
        """
        Returns continuous probability map P_MedSAM-LoRA(x, y) in [0.0, 1.0] of shape (512, 512),
        taking pixel-wise max probability across all K candidate boxes.
        """
        if len(boxes_list) == 0:
            return np.zeros(TARGET_SHAPE, dtype=np.float32)

        K = len(boxes_list)
        boxes_t = torch.tensor(boxes_list, dtype=torch.float32, device=device).unsqueeze(1)  # (K, 1, 4)
        sparse_emb, dense_emb = self.sam.prompt_encoder(points=None, boxes=boxes_t, masks=None)

        dec = self.sam.mask_decoder
        dense_pe = self.sam.prompt_encoder.get_dense_pe()
        output_tokens = torch.cat([dec.iou_token.weight, dec.mask_tokens.weight], dim=0).unsqueeze(0).expand(K, -1, -1)
        tokens = torch.cat((output_tokens, sparse_emb), dim=1)

        src = image_embeddings.expand(K, -1, -1, -1) + dense_emb
        pos_src = dense_pe.expand(K, -1, -1, -1)
        b, c, h, w = src.shape

        hs, src = dec.transformer(src, pos_src, tokens)
        mask_tokens_out = hs[:, 1:2, :]
        src = src.transpose(1, 2).view(b, c, h, w)
        upscaled = dec.output_upscaling(src)
        hyper_in = dec.output_hypernetworks_mlps[0](mask_tokens_out[:, 0, :]).unsqueeze(1)

        b, c_up, h_up, w_up = upscaled.shape
        low_res = (hyper_in @ upscaled.view(b, c_up, h_up * w_up)).view(b, 1, h_up, w_up)
        up_logits = F.interpolate(low_res.float(), size=TARGET_SHAPE, mode="bilinear", align_corners=False)

        prob_per_box = torch.sigmoid(up_logits).squeeze(1)  # (K, 512, 512)
        merged_prob = prob_per_box.max(dim=0).values
        return merged_prob.cpu().numpy().astype(np.float32)

medsam_lora = MedSAMLoRABinary(MEDSAM_BASE_CKPT, rank=8, alpha=16)
medsam_lora.load_state_dict(extract_clean_state_dict(MEDSAM_LORA_CKPT))
medsam_lora = medsam_lora.to(device)
medsam_lora.eval()
print(f"[+] Stage 2 Loaded: {MEDSAM_LORA_CKPT} | BOX_PADDING = {BOX_PADDING} px")

# =====================================================================
# 3. Load & Cache Hold-Out Test Split (CLAHE + 512x512)
# =====================================================================
json_candidates = list((DATASET_ROOT / "Annotations").rglob("*COCO*.json")) or list((DATASET_ROOT / "Annotations").rglob("*.json"))
with open(json_candidates[0], "r") as f:
    coco_payload = json.load(f)

img_id_to_annos = {}
for ann in coco_payload["annotations"]:
    img_id_to_annos.setdefault(ann["image_id"], []).append(ann)

test_df = pd.read_csv(TEST_CSV).reset_index(drop=True)

def _load_item(row_dict):
    clahe = cv2.createCLAHE(clipLimit=2.0, tileGridSize=(8, 8))
    gray = cv2.imread(str(row_dict["path"]), cv2.IMREAD_GRAYSCALE)
    orig_h, orig_w = gray.shape
    resized = cv2.resize(clahe.apply(gray), TARGET_SHAPE, interpolation=cv2.INTER_CUBIC)
    norm = (resized.astype(np.float32) - resized.min()) / (resized.max() - resized.min() + 1e-8)

    gt_nat = np.zeros((orig_h, orig_w), dtype=np.uint8)
    coco_id = int(row_dict["coco_id"])
    if coco_id != -1 and coco_id in img_id_to_annos:
        for ann in img_id_to_annos[coco_id]:
            for poly in ann.get("segmentation", []):
                pts = np.array(poly, dtype=np.int32).reshape((-1, 2))
                cv2.fillPoly(gt_nat, [pts], 1)

    gt_512 = cv2.resize(gt_nat, TARGET_SHAPE, interpolation=cv2.INTER_NEAREST)
    return {
        "file_name": row_dict["file_name"],
        "fractured": int(row_dict["fractured"]),
        "resized_u8": resized,
        "norm_f32": norm,
        "gt_512": gt_512,
    }

with ThreadPoolExecutor(max_workers=min(8, os.cpu_count() or 4)) as ex:
    cached_test = list(ex.map(_load_item, test_df.to_dict(orient="records")))

# =====================================================================
# 4. Stage-1 Probability Extraction, Box ROI Mask & Metric Utilities
# =====================================================================
@torch.inference_mode()
def get_stage1_probs(model, inp_512):
    """
    Returns:
      - p_frac_1v: (512, 512) float32 fracture probability map (1-view)
      - mask_1v:   (512, 512) uint8 discrete argmax fracture mask (1-view)
      - p_frac_4v: (512, 512) float32 fracture probability map (4-view TTA)
      - mask_4v:   (512, 512) uint8 discrete argmax fracture mask (4-view TTA)
    """
    inp_flip = torch.flip(inp_512, dims=[3])
    probs_2 = torch.softmax(model(torch.cat([inp_512, inp_flip], dim=0)), dim=1)
    p1 = probs_2[0:1]
    p2 = torch.flip(probs_2[1:2], dims=[3])

    inp_480 = F.interpolate(inp_512, size=(480, 480), mode="bilinear", align_corners=False)
    p3 = F.interpolate(torch.softmax(model(inp_480), dim=1), size=TARGET_SHAPE, mode="bilinear", align_corners=False)

    inp_544 = F.interpolate(inp_512, size=(544, 544), mode="bilinear", align_corners=False)
    p4 = F.interpolate(torch.softmax(model(inp_544), dim=1), size=TARGET_SHAPE, mode="bilinear", align_corners=False)

    p_avg_4 = 0.25 * (p1 + p2 + p3 + p4)

    p_frac_1v = p1[0, 2].cpu().numpy().astype(np.float32)
    mask_1v = (torch.argmax(p1, dim=1).squeeze(0) == 2).cpu().numpy().astype(np.uint8)

    p_frac_4v = p_avg_4[0, 2].cpu().numpy().astype(np.float32)
    mask_4v = (torch.argmax(p_avg_4, dim=1).squeeze(0) == 2).cpu().numpy().astype(np.uint8)

    return p_frac_1v, mask_1v, p_frac_4v, mask_4v

def extract_boxes_and_roi(bin_mask, pad=BOX_PADDING, min_area=MIN_BLOB_AREA):
    H, W = bin_mask.shape
    num_labels, _, stats, _ = cv2.connectedComponentsWithStats(bin_mask.astype(np.uint8), connectivity=8)
    boxes = []
    roi_mask = np.zeros((H, W), dtype=np.float32)
    for lbl in range(1, num_labels):
        if stats[lbl, cv2.CC_STAT_AREA] < min_area:
            continue
        x, y, w, h = stats[lbl, cv2.CC_STAT_LEFT], stats[lbl, cv2.CC_STAT_TOP], stats[lbl, cv2.CC_STAT_WIDTH], stats[lbl, cv2.CC_STAT_HEIGHT]
        x1, y1 = max(0, x - pad), max(0, y - pad)
        x2, y2 = min(W - 1, x + w + pad), min(H - 1, y + h + pad)
        boxes.append([x1, y1, x2, y2])
        roi_mask[y1 : y2 + 1, x1 : x2 + 1] = 1.0
    return boxes, roi_mask

DIAG_PENALTY = float(np.hypot(TARGET_SHAPE[0], TARGET_SHAPE[1]))
BOUNDARY_KERNEL = np.ones((3, 3), dtype=np.uint8)

def compute_metrics(pred_bin, gt_bin, smooth=1e-5):
    p = (pred_bin > 0).astype(np.uint8)
    g = (gt_bin > 0).astype(np.uint8)
    inter = int(np.logical_and(p, g).sum())
    p_sum, g_sum = int(p.sum()), int(g.sum())
    union = int(np.logical_or(p, g).sum())

    dice = (2.0 * inter + smooth) / (p_sum + g_sum + smooth)
    iou = (inter + smooth) / (union + smooth)

    if p_sum == 0 and g_sum == 0:
        hd95 = 0.0
    elif p_sum == 0 or g_sum == 0:
        hd95 = DIAG_PENALTY
    else:
        p_bnd = p - cv2.erode(p, BOUNDARY_KERNEL, iterations=1)
        g_bnd = g - cv2.erode(g, BOUNDARY_KERNEL, iterations=1)
        if p_bnd.sum() == 0:
            p_bnd = p
        if g_bnd.sum() == 0:
            g_bnd = g
        dt_g = cv2.distanceTransform(1 - g_bnd, cv2.DIST_L2, 5)
        dt_p = cv2.distanceTransform(1 - p_bnd, cv2.DIST_L2, 5)
        dists = np.concatenate([dt_g[p_bnd > 0], dt_p[g_bnd > 0]])
        hd95 = float(np.percentile(dists, 95))

    return float(dice), float(iou), hd95, inter, p_sum, g_sum

# =====================================================================
# 5. Cache Continuous Probability Maps Once & Sweep Alpha Fusion
# =====================================================================
cached_probs = []
with torch.inference_mode():
    for item in tqdm(cached_test, desc="Extracting Stage-1 & Stage-2 Probability Maps"):
        inp = torch.from_numpy(item["norm_f32"]).unsqueeze(0).unsqueeze(0).repeat(1, 3, 1, 1).to(device)
        p_u1, m_u1, p_u4, m_u4 = get_stage1_probs(unetpp, inp)

        boxes_1v, roi_1v = extract_boxes_and_roi(m_u1)
        boxes_4v, roi_4v = extract_boxes_and_roi(m_u4)

        with torch.amp.autocast("cuda", dtype=amp_dtype, enabled=(device.type == "cuda")):
            img_embed = medsam_lora.sam.image_encoder(inp.to(dtype=amp_dtype))
            p_sam_1v = medsam_lora.decode_boxes_prob_map(img_embed, boxes_1v)
            p_sam_4v = medsam_lora.decode_boxes_prob_map(img_embed, boxes_4v)

        cached_probs.append({
            "p_u1": p_u1, "roi_1v": roi_1v, "p_sam_1v": p_sam_1v, "boxes_1v": boxes_1v,
            "p_u4": p_u4, "roi_4v": roi_4v, "p_sam_4v": p_sam_4v, "boxes_4v": boxes_4v,
        })

benchmark_rows = []
vis_cache = {}  # key -> list of binary masks aligned with cached_test

for s1_mode in ["1-View Stage 1", "4-View TTA Stage 1"]:
    for alpha in ALPHAS:
        dices, ious, hd95s, healthy_fps = [], [], [], []
        m_inter, m_pred, m_gt = 0, 0, 0
        preds_for_mode = []

        for item, cp in zip(cached_test, cached_probs):
            if s1_mode == "1-View Stage 1":
                p_unet, roi, p_sam = cp["p_u1"], cp["roi_1v"], cp["p_sam_1v"]
            else:
                p_unet, roi, p_sam = cp["p_u4"], cp["roi_4v"], cp["p_sam_4v"]

            # Confidence-weighted probability fusion strictly inside Stage-1 boxes
            p_cascade = roi * (alpha * p_sam + (1.0 - alpha) * p_unet)
            pred_bin = (p_cascade > PROB_THRESH).astype(np.uint8)
            preds_for_mode.append(pred_bin)

            if item["fractured"] == 1:
                d, iou, h95, inter, p_s, g_s = compute_metrics(pred_bin, item["gt_512"])
                dices.append(d); ious.append(iou); hd95s.append(h95)
                m_inter += inter; m_pred += p_s; m_gt += g_s
            else:
                healthy_fps.append(float(pred_bin.sum()))

        micro_dice = (2.0 * m_inter + 1e-5) / (m_pred + m_gt + 1e-5)
        fusion_tag = "Hard Stage 2 Only (alpha=1.00)" if alpha == 1.0 else f"Soft Blend (alpha={alpha:.2f})"
        benchmark_rows.append({
            "Stage-1 Proposal": s1_mode,
            "Probability Fusion": fusion_tag,
            "Macro-Dice": round(float(np.mean(dices)), 4),
            "Micro-Dice": round(float(micro_dice), 4),
            "Macro-IoU": round(float(np.mean(ious)), 4),
            "HD95 (px)": round(float(np.mean(hd95s)), 2),
            "Healthy FP (px)": round(float(np.mean(healthy_fps)), 2),
        })
        vis_cache[(s1_mode, alpha)] = preds_for_mode

df_res = pd.DataFrame(benchmark_rows)
print("\n" + "=" * 116)
print(f"CONFIDENCE-WEIGHTED PROBABILITY FUSION ABLATION (BOX_PADDING = {BOX_PADDING} px | FracAtlas Test N = 216 | SEED = 42)")
print("=" * 116)
print(df_res.to_string(index=False))
print("=" * 116)

df_res.to_csv(OUT_CSV_PATH, index=False)
print(f"[+] Saved CSV results to: {OUT_CSV_PATH}")

# =====================================================================
# 6. Save 20-Sample Visual Comparison (Hard alpha=1.00 vs Best Soft alpha=0.65)
# =====================================================================
frac_indices = [i for i, r in enumerate(cached_test) if r["fractured"] == 1]
rng = random.Random(SEED)
sample_20 = rng.sample(frac_indices, k=min(20, len(frac_indices)))

def overlay_mask(gray_u8, mask_bin, rgb_color, alpha=0.58):
    canvas = cv2.cvtColor(gray_u8, cv2.COLOR_GRAY2RGB).astype(np.float32)
    fg = mask_bin > 0
    col = np.array(rgb_color, dtype=np.float32)
    canvas[fg] = (1.0 - alpha) * canvas[fg] + alpha * col
    return np.clip(canvas, 0, 255).astype(np.uint8)

fig, axes = plt.subplots(nrows=len(sample_20), ncols=4, figsize=(16, 3.8 * len(sample_20)))
for row_i, g_idx in enumerate(sample_20):
    item = cached_test[g_idx]
    cp = cached_probs[g_idx]
    gt = item["gt_512"]

    hard_1v = vis_cache[("1-View Stage 1", 1.00)][g_idx]
    soft_1v = vis_cache[("1-View Stage 1", 0.65)][g_idx]
    soft_4v = vis_cache[("4-View TTA Stage 1", 0.65)][g_idx]

    d_h1, _, _, _, _, _ = compute_metrics(hard_1v, gt)
    d_s1, _, _, _, _, _ = compute_metrics(soft_1v, gt)
    d_s4, _, _, _, _, _ = compute_metrics(soft_4v, gt)

    gt_vis = overlay_mask(item["resized_u8"], gt, (255, 30, 30))
    h1_vis = overlay_mask(item["resized_u8"], hard_1v, (40, 170, 255))
    s1_vis = overlay_mask(item["resized_u8"], soft_1v, (255, 190, 30))
    s4_vis = overlay_mask(item["resized_u8"], soft_4v, (40, 240, 40))

    for (x1, y1, x2, y2) in cp["boxes_4v"]:
        cv2.rectangle(s4_vis, (x1, y1), (x2, y2), (255, 225, 0), 2)

    axes[row_i, 0].imshow(gt_vis)
    axes[row_i, 0].set_title(f"[{row_i+1:02d}] GT | {item['file_name']}", fontsize=9, color="#b30000", fontweight="bold")
    axes[row_i, 0].axis("off")

    axes[row_i, 1].imshow(h1_vis)
    axes[row_i, 1].set_title(f"Hard Cascade (a=1.0) | Dice: {d_h1:.3f}", fontsize=9, color="#005599")
    axes[row_i, 1].axis("off")

    axes[row_i, 2].imshow(s1_vis)
    axes[row_i, 2].set_title(f"1-View Soft Fusion (a=0.65) | Dice: {d_s1:.3f}", fontsize=9, color="#995500")
    axes[row_i, 2].axis("off")

    axes[row_i, 3].imshow(s4_vis)
    axes[row_i, 3].set_title(f"4-View Soft Fusion (a=0.65) | Dice: {d_s4:.3f}", fontsize=9, color="#006600", fontweight="bold")
    axes[row_i, 3].axis("off")

plt.tight_layout()
plt.savefig(OUT_FIG_PATH, dpi=100, bbox_inches="tight")
print(f"[+] Saved 20-scan probability fusion comparison grid to: {OUT_FIG_PATH}")
```

```python
import os
os.environ["OPENCV_LOG_LEVEL"] = "SILENT"
os.environ["PYTORCH_CUDA_ALLOC_CONF"] = "expandable_segments:True"

import json
import time
import random
from pathlib import Path

import cv2
import numpy as np
import pandas as pd
import matplotlib
matplotlib.use("Agg")
import matplotlib.pyplot as plt
from tqdm import tqdm

import torch
import torch.nn as nn
import torch.nn.functional as F
import segmentation_models_pytorch as smp
from segmentation_models_pytorch.base import SegmentationHead

# =====================================================================
# 1. Global Constants & RTX 4050 Setup (Locked 512 x 512, SEED = 42)
# =====================================================================
SEED = 42
TARGET_SHAPE = (512, 512)
BATCH_SIZE = 4              # Micro-batch size for 6GB RTX 4050 VRAM
ACCUM_STEPS = 2             # Effective batch size = 8
EPOCHS = 40
LR_ENCODER = 1.5e-4
LR_DECODER = 4.0e-4

def seed_everything(seed=42):
    random.seed(seed)
    os.environ["PYTHONHASHSEED"] = str(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)
    torch.cuda.manual_seed_all(seed)
    torch.backends.cudnn.benchmark = True
    torch.set_float32_matmul_precision("high")

seed_everything(SEED)
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
use_bf16 = torch.cuda.is_available() and torch.cuda.is_bf16_supported()
amp_dtype = torch.bfloat16 if use_bf16 else torch.float16
print(f"[+] Device: {device} ({torch.cuda.get_device_name(0) if torch.cuda.is_available() else 'CPU'}) | 512x512 | AMP: {amp_dtype}")

# =====================================================================
# 2. Paths & New Descriptive Output Directory
# =====================================================================
DATASET_ROOT = Path("./FracAtlas")
SPLITS_DIR = Path("./splits")
SAVE_DIR = Path("./FracAtlas_UNetPP_EfficientNetB3_DeepSup")
SAVE_DIR.mkdir(parents=True, exist_ok=True)

BEST_CKPT_PATH = SAVE_DIR / "unetpp_effnetb3_deepsup_best.pt"
HISTORY_CSV_PATH = SAVE_DIR / "training_history.csv"
CURVES_FIG_PATH = SAVE_DIR / "training_curves.png"
TEST_SUMMARY_CSV = SAVE_DIR / "test_evaluation_summary.csv"
TEST_VIS_FIG_PATH = SAVE_DIR / "test_predictions_20_comparison.png"

train_csv = SPLITS_DIR / "train_split_50_50.csv"
val_csv   = SPLITS_DIR / "val_split_50_50.csv"
test_csv  = SPLITS_DIR / "test_split_50_50.csv"

for p in [train_csv, val_csv, test_csv]:
    if not p.exists():
        raise FileNotFoundError(f"Required split CSV not found: {p}")

json_candidates = list((DATASET_ROOT / "Annotations").rglob("*COCO*.json")) or list((DATASET_ROOT / "Annotations").rglob("*.json"))
with open(json_candidates[0], "r") as f:
    coco_payload = json.load(f)

img_id_to_annos = {}
for ann in coco_payload["annotations"]:
    img_id_to_annos.setdefault(ann["image_id"], []).append(ann)

# =====================================================================
# 3. Preload 512x512 Dataset into Pinned System RAM (Zero Disk Bottleneck)
# =====================================================================
clahe = cv2.createCLAHE(clipLimit=2.0, tileGridSize=(8, 8))

def load_split_to_pinned_ram(csv_path, split_name):
    df = pd.read_csv(csv_path).reset_index(drop=True)
    N = len(df)
    H, W = TARGET_SHAPE

    imgs_np = np.empty((N, 1, H, W), dtype=np.uint8)
    masks_np = np.empty((N, H, W), dtype=np.uint8)
    frac_flags = df["fractured"].astype(np.int64).values
    file_names = df["file_name"].tolist()

    for i, row in tqdm(df.iterrows(), total=N, desc=f"Caching {split_name} -> RAM"):
        gray = cv2.imread(str(row["path"]), cv2.IMREAD_GRAYSCALE)
        if gray is None:
            raise FileNotFoundError(f"Cannot read image: {row['path']}")
        orig_h, orig_w = gray.shape

        enhanced = clahe.apply(gray)
        resized = cv2.resize(enhanced, (W, H), interpolation=cv2.INTER_CUBIC)
        r_min, r_max = float(resized.min()), float(resized.max())
        imgs_np[i, 0] = ((resized.astype(np.float32) - r_min) / (r_max - r_min + 1e-8) * 255.0).astype(np.uint8)

        # Class 1: Intact Bone pseudo-label (Otsu + morphological closing)
        blurred = cv2.GaussianBlur(resized, (7, 7), 1.5)
        _, binary = cv2.threshold(blurred, 0, 255, cv2.THRESH_BINARY + cv2.THRESH_OTSU)
        kernel = cv2.getStructuringElement(cv2.MORPH_ELLIPSE, (9, 9))
        closed = cv2.morphologyEx(binary, cv2.MORPH_CLOSE, kernel, iterations=2)
        num_labels, labels, stats, _ = cv2.connectedComponentsWithStats(closed, connectivity=8)
        bone_mask = np.zeros_like(closed, dtype=np.uint8)
        for lbl in range(1, num_labels):
            if stats[lbl, cv2.CC_STAT_AREA] >= 450:
                bone_mask[labels == lbl] = 1

        # Class 2: Verified Fracture polygons from COCO JSON
        frac_native = np.zeros((orig_h, orig_w), dtype=np.uint8)
        coco_id = int(row["coco_id"])
        if coco_id != -1 and coco_id in img_id_to_annos:
            for ann in img_id_to_annos[coco_id]:
                seg = ann.get("segmentation", [])
                if isinstance(seg, list):
                    for poly in seg:
                        pts = np.array(poly, dtype=np.int32).reshape((-1, 2))
                        cv2.fillPoly(frac_native, [pts], color=1)

        frac_resized = cv2.resize(frac_native, (W, H), interpolation=cv2.INTER_NEAREST)

        m3 = np.zeros((H, W), dtype=np.uint8)
        m3[bone_mask > 0] = 1
        m3[frac_resized > 0] = 2
        masks_np[i] = m3

    imgs_t = torch.from_numpy(imgs_np)
    masks_t = torch.from_numpy(masks_np)
    flags_t = torch.from_numpy(frac_flags)
    if device.type == "cuda":
        imgs_t = imgs_t.pin_memory()
        masks_t = masks_t.pin_memory()
    return imgs_t, masks_t, flags_t, file_names

train_imgs, train_masks, train_flags, _ = load_split_to_pinned_ram(train_csv, "Train (70%)")
val_imgs,   val_masks,   val_flags,   _ = load_split_to_pinned_ram(val_csv,   "Val   (15%)")
test_imgs,  test_masks,  test_flags,  test_names = load_split_to_pinned_ram(test_csv, "Test  (15%)")

# =====================================================================
# 4. U-Net++ (EfficientNet-B3) + Auto-Sized Deep Supervision Wrapper
# =====================================================================
class UNetPlusPlusDeepSupervision(nn.Module):
    def __init__(self, encoder_name="efficientnet-b3", encoder_weights="imagenet", classes=3, input_size=512):
        super().__init__()
        self.net = smp.UnetPlusPlus(
            encoder_name=encoder_name,
            encoder_weights=encoder_weights,
            in_channels=3,
            classes=classes,
            decoder_channels=(256, 128, 64, 32, 16),
        )
        top_keys = sorted(
            [k for k in self.net.decoder.blocks.keys() if k.startswith("x_0_") and k != "x_0_0"],
            key=lambda s: int(s.split("_")[-1])
        )
        self.aux_keys = top_keys[:-1]
        self._aux_feats = {}
        self._capture_for_init = True

        for k in self.aux_keys:
            self.net.decoder.blocks[k].register_forward_hook(self._make_hook(k))

        with torch.no_grad():
            _ = self.net(torch.zeros(1, 3, input_size, input_size))

        self.aux_heads = nn.ModuleDict()
        for k in self.aux_keys:
            feat = self._aux_feats[k]
            c_in = feat.shape[1]
            up_scale = input_size // feat.shape[2]
            self.aux_heads[k] = SegmentationHead(
                in_channels=c_in,
                out_channels=classes,
                kernel_size=3,
                upsampling=up_scale,
            )

        self._aux_feats.clear()
        self._capture_for_init = False

    def _make_hook(self, key):
        def hook(_module, _inp, output):
            if self.training or self._capture_for_init:
                self._aux_feats[key] = output
        return hook

    def forward(self, x):
        self._aux_feats.clear()
        final_logits = self.net(x)
        if self.training:
            outputs = [self.aux_heads[k](self._aux_feats[k]) for k in self.aux_keys]
            outputs.append(final_logits)
            self._aux_feats.clear()
            return outputs
        return final_logits

model = UNetPlusPlusDeepSupervision(
    encoder_name="efficientnet-b3",
    encoder_weights="imagenet",
    classes=3,
    input_size=TARGET_SHAPE[0],
).to(device=device, memory_format=torch.channels_last)

total_params = sum(p.numel() for p in model.parameters())
print(f"[+] Built U-Net++ (EfficientNet-B3 + Deep Supervision) | Total Params: {total_params / 1e6:.2f}M")

# =====================================================================
# 5. Hybrid Focal-Tversky + Weighted Cross-Entropy Deep Supervision Loss
# =====================================================================
class_weights = torch.tensor([0.1, 0.3, 1.0], device=device)
alpha_vec = torch.tensor([0.5, 0.5, 0.7], device=device)
beta_vec  = torch.tensor([0.5, 0.5, 0.3], device=device)
criterion_ce = nn.CrossEntropyLoss(weight=class_weights)
DS_WEIGHTS = [0.20, 0.30, 0.50, 1.00]

def single_head_hybrid_loss(logits, targets, gamma=1.33, smooth=1e-5):
    ce = criterion_ce(logits, targets)
    probs = F.softmax(logits.float(), dim=1)
    targets_oh = F.one_hot(targets, num_classes=3).permute(0, 3, 1, 2).float()

    dims = (0, 2, 3)
    tp = torch.sum(probs * targets_oh, dims)
    fn = torch.sum((1.0 - probs) * targets_oh, dims)
    fp = torch.sum(probs * (1.0 - targets_oh), dims)

    tversky_index = (tp + smooth) / (tp + alpha_vec * fn + beta_vec * fp + smooth)
    focal_tversky = torch.pow(1.0 - tversky_index, gamma)
    ft_loss = torch.sum(class_weights * focal_tversky) / torch.sum(class_weights)
    return ce + ft_loss

def deep_supervision_loss(outputs_list, targets):
    total = 0.0
    norm_factor = sum(DS_WEIGHTS)
    for w, head_logits in zip(DS_WEIGHTS, outputs_list):
        total = total + (w / norm_factor) * single_head_hybrid_loss(head_logits, targets)
    return total

def compute_batch_dice(logits, targets, class_idx=2, smooth=1e-5):
    preds = torch.argmax(logits, dim=1)
    pred_c = (preds == class_idx).float()
    targ_c = (targets == class_idx).float()
    inter = (pred_c * targ_c).sum()
    return ((2.0 * inter + smooth) / (pred_c.sum() + targ_c.sum() + smooth)).item()

optimizer = torch.optim.AdamW(
    [
        {"params": model.net.encoder.parameters(), "lr": LR_ENCODER},
        {"params": model.net.decoder.parameters(), "lr": LR_DECODER},
        {"params": model.net.segmentation_head.parameters(), "lr": LR_DECODER},
        {"params": model.aux_heads.parameters(), "lr": LR_DECODER},
    ],
    weight_decay=1e-4,
    fused=(device.type == "cuda"),
)
scheduler = torch.optim.lr_scheduler.CosineAnnealingLR(optimizer, T_max=EPOCHS, eta_min=1e-6)
scaler = torch.amp.GradScaler("cuda", enabled=(device.type == "cuda" and not use_bf16))

# =====================================================================
# 6. Training Loop (40 Epochs @ 512x512, Saving Best Validation Checkpoint)
# =====================================================================
best_val_frac_dice = 0.0
n_train = train_imgs.size(0)
n_val = val_imgs.size(0)
history = []

print(f"\n[*] Starting Training -> {BEST_CKPT_PATH}\n")
t_train_start = time.time()

for epoch in range(1, EPOCHS + 1):
    ep_t0 = time.time()
    model.train()
    running_loss = torch.zeros(1, device=device)
    num_batches = 0

    perm = torch.randperm(n_train)
    total_steps = (n_train + BATCH_SIZE - 1) // BATCH_SIZE
    optimizer.zero_grad(set_to_none=True)

    train_pbar = tqdm(
        range(0, n_train, BATCH_SIZE),
        total=total_steps,
        desc=f"Epoch {epoch:02d}/{EPOCHS:02d} [Train]",
        leave=False,
        mininterval=0.5,
    )

    for step_idx, start in enumerate(train_pbar):
        idx = perm[start : start + BATCH_SIZE]
        imgs = train_imgs[idx].to(device, non_blocking=True).to(dtype=amp_dtype).div_(255.0)
        masks = train_masks[idx].to(device, non_blocking=True).long()

        if random.random() > 0.5:
            imgs = torch.flip(imgs, dims=[3])
            masks = torch.flip(masks, dims=[2])
        if random.random() > 0.5:
            scale_jitter = random.uniform(0.90, 1.10)
            imgs = torch.clamp(imgs * scale_jitter, 0.0, 1.0)

        imgs = imgs.repeat(1, 3, 1, 1).to(memory_format=torch.channels_last)

        with torch.amp.autocast("cuda", dtype=amp_dtype, enabled=(device.type == "cuda")):
            outputs_list = model(imgs)
            loss = deep_supervision_loss(outputs_list, masks)
            loss_scaled = loss / ACCUM_STEPS

        if scaler.is_enabled():
            scaler.scale(loss_scaled).backward()
        else:
            loss_scaled.backward()

        if ((step_idx + 1) % ACCUM_STEPS == 0) or ((step_idx + 1) == total_steps):
            if scaler.is_enabled():
                scaler.step(optimizer)
                scaler.update()
            else:
                optimizer.step()
            optimizer.zero_grad(set_to_none=True)

        running_loss += loss.detach()
        num_batches += 1

    scheduler.step()
    avg_train_loss = (running_loss / num_batches).item()

    model.eval()
    val_loss_t = torch.zeros(1, device=device)
    bone_dices, frac_dices = [], []
    val_batches = 0

    with torch.inference_mode():
        for start in range(0, n_val, BATCH_SIZE):
            imgs = val_imgs[start : start + BATCH_SIZE].to(device, non_blocking=True).to(dtype=amp_dtype).div_(255.0)
            imgs = imgs.repeat(1, 3, 1, 1).to(memory_format=torch.channels_last)
            masks = val_masks[start : start + BATCH_SIZE].to(device, non_blocking=True).long()
            flags = val_flags[start : start + BATCH_SIZE]

            with torch.amp.autocast("cuda", dtype=amp_dtype, enabled=(device.type == "cuda")):
                logits = model(imgs)
                loss = single_head_hybrid_loss(logits, masks)

            val_loss_t += loss.detach()
            val_batches += 1
            bone_dices.append(compute_batch_dice(logits, masks, class_idx=1))
            if flags.sum() > 0:
                f_idx = (flags == 1).to(device)
                frac_dices.append(compute_batch_dice(logits[f_idx], masks[f_idx], class_idx=2))

    avg_val_loss = (val_loss_t / val_batches).item()
    mean_bone_dice = float(np.mean(bone_dices))
    mean_frac_dice = float(np.mean(frac_dices)) if frac_dices else 0.0
    ep_sec = time.time() - ep_t0

    history.append({
        "epoch": epoch,
        "train_loss": round(avg_train_loss, 5),
        "val_loss": round(avg_val_loss, 5),
        "val_bone_dice": round(mean_bone_dice, 4),
        "val_frac_dice": round(mean_frac_dice, 4),
        "epoch_sec": round(ep_sec, 1),
    })

    print(f"Epoch {epoch:02d}/{EPOCHS:02d} ({ep_sec:.1f}s) | Train Loss: {avg_train_loss:.4f} | "
          f"Val Loss: {avg_val_loss:.4f} | Bone Dice: {mean_bone_dice:.4f} | Frac Dice: {mean_frac_dice:.4f}")

    if mean_frac_dice > best_val_frac_dice:
        best_val_frac_dice = mean_frac_dice
        torch.save({
            "epoch": epoch,
            "encoder": "efficientnet-b3",
            "deep_supervision": True,
            "best_val_frac_dice": best_val_frac_dice,
            "model_state_dict": model.state_dict(),
        }, BEST_CKPT_PATH)
        print(f"   --> Saved Best Model (Val Fracture Dice: {best_val_frac_dice:.4f})")

print(f"\n[+] Finished {EPOCHS} epochs in {(time.time() - t_train_start)/60:.2f} minutes!")

df_hist = pd.DataFrame(history)
df_hist.to_csv(HISTORY_CSV_PATH, index=False)

fig_c, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 4.5))
ax1.plot(df_hist["epoch"], df_hist["train_loss"], label="Train Deep-Sup Loss", linewidth=2)
ax1.plot(df_hist["epoch"], df_hist["val_loss"], label="Val Hybrid Loss", linewidth=2)
ax1.set_xlabel("Epoch"); ax1.set_ylabel("Loss"); ax1.set_title("U-Net++ (EfficientNet-B3 + DS) Loss")
ax1.legend(); ax1.grid(True, alpha=0.3)

ax2.plot(df_hist["epoch"], df_hist["val_frac_dice"], color="green", label="Val Fracture Dice", linewidth=2)
ax2.plot(df_hist["epoch"], df_hist["val_bone_dice"], color="steelblue", linestyle="--", label="Val Bone Dice", linewidth=1.8)
ax2.set_xlabel("Epoch"); ax2.set_ylabel("Dice Score"); ax2.set_title("Validation Dice Progression")
ax2.legend(); ax2.grid(True, alpha=0.3)

plt.tight_layout()
plt.savefig(CURVES_FIG_PATH, dpi=120)
plt.close(fig_c)

# =====================================================================
# 7. Hold-Out Test Split Evaluation (N = 216, Macro/Micro Dice, IoU, HD95)
# =====================================================================
ckpt = torch.load(BEST_CKPT_PATH, map_location=device)
model.load_state_dict(ckpt["model_state_dict"])
model = model.float().to(device)
model.eval()

DIAG_PENALTY = float(np.hypot(TARGET_SHAPE[0], TARGET_SHAPE[1]))
BOUNDARY_KERNEL = np.ones((3, 3), dtype=np.uint8)

def compute_scan_metrics(pred_bin, gt_bin, smooth=1e-5):
    p = (pred_bin > 0).astype(np.uint8)
    g = (gt_bin > 0).astype(np.uint8)
    inter = int(np.logical_and(p, g).sum())
    p_sum, g_sum = int(p.sum()), int(g.sum())
    union = int(np.logical_or(p, g).sum())

    dice = (2.0 * inter + smooth) / (p_sum + g_sum + smooth)
    iou = (inter + smooth) / (union + smooth)

    if p_sum == 0 and g_sum == 0:
        hd95 = 0.0
    elif p_sum == 0 or g_sum == 0:
        hd95 = DIAG_PENALTY
    else:
        p_bnd = p - cv2.erode(p, BOUNDARY_KERNEL, iterations=1)
        g_bnd = g - cv2.erode(g, BOUNDARY_KERNEL, iterations=1)
        if p_bnd.sum() == 0:
            p_bnd = p
        if g_bnd.sum() == 0:
            g_bnd = g
        dt_g = cv2.distanceTransform(1 - g_bnd, cv2.DIST_L2, 5)
        dt_p = cv2.distanceTransform(1 - p_bnd, cv2.DIST_L2, 5)
        dists = np.concatenate([dt_g[p_bnd > 0], dt_p[g_bnd > 0]])
        hd95 = float(np.percentile(dists, 95))

    return float(dice), float(iou), hd95, inter, p_sum, g_sum

macro_dices, macro_ious, macro_hd95s = [], [], []
bone_dices, healthy_fps, latencies_ms = [], [], []
micro_inter, micro_pred, micro_gt = 0, 0, 0
detected_scans = 0
vis_records = []

with torch.inference_mode():
    for i in tqdm(range(test_imgs.size(0)), desc="Evaluating Hold-Out Test Split (N=216)"):
        t0 = time.time()
        inp = test_imgs[i : i + 1].to(device).float().div_(255.0).repeat(1, 3, 1, 1)
        logits = model(inp)
        pred_cls = torch.argmax(logits, dim=1).squeeze(0).cpu().numpy().astype(np.uint8)
        if device.type == "cuda":
            torch.cuda.synchronize()
        latencies_ms.append((time.time() - t0) * 1000.0)

        gt_cls = test_masks[i].numpy()
        b_dice, _, _, _, _, _ = compute_scan_metrics((pred_cls == 1), (gt_cls == 1))
        bone_dices.append(b_dice)

        pred_frac = (pred_cls == 2).astype(np.uint8)
        gt_frac = (gt_cls == 2).astype(np.uint8)

        if int(test_flags[i].item()) == 1:
            d, iou, h95, inter, p_s, g_s = compute_scan_metrics(pred_frac, gt_frac)
            macro_dices.append(d); macro_ious.append(iou); macro_hd95s.append(h95)
            micro_inter += inter; micro_pred += p_s; micro_gt += g_s
            if p_s > 0:
                detected_scans += 1

            vis_records.append({
                "file_name": test_names[i],
                "img": test_imgs[i, 0].numpy(),
                "gt": gt_frac,
                "pred": pred_frac,
                "dice": d,
            })
        else:
            healthy_fps.append(float(pred_frac.sum()))

micro_dice = (2.0 * micro_inter + 1e-5) / (micro_pred + micro_gt + 1e-5)
scan_recall_pct = detected_scans / max(1, len(macro_dices)) * 100.0

summary = {
    "Model": "U-Net++ (EfficientNet-B3 + Deep Supervision)",
    "Checkpoint": str(BEST_CKPT_PATH),
    "Best Epoch": int(ckpt["epoch"]),
    "Total Params (M)": round(total_params / 1e6, 2),
    "Scan-Level Recall (>0 px)": f"{detected_scans}/{len(macro_dices)} ({scan_recall_pct:.1f}%)",
    "Fracture Macro-Dice": round(float(np.mean(macro_dices)), 4),
    "Fracture Micro-Dice": round(float(micro_dice), 4),
    "Fracture Macro-IoU": round(float(np.mean(macro_ious)), 4),
    "Boundary HD95 (px)": round(float(np.mean(macro_hd95s)), 2),
    "Bone Anatomy Dice": round(float(np.mean(bone_dices)), 4),
    "Healthy FP (px/scan)": round(float(np.mean(healthy_fps)), 2),
    "Latency (ms/scan)": round(float(np.mean(latencies_ms)), 1),
}

pd.DataFrame([summary]).to_csv(TEST_SUMMARY_CSV, index=False)

print("\n======== U-Net++ (EfficientNet-B3 + Deep Supervision) TEST RESULTS ========")
for k, v in summary.items():
    print(f"{k:<30}: {v}")
print("===========================================================================")

# =====================================================================
# 8. Save 20-Scan Qualitative Overlay Grid
# =====================================================================
rng = random.Random(SEED)
sample_20 = rng.sample(vis_records, k=min(20, len(vis_records)))

def overlay_mask(gray_u8, mask_bin, rgb_color, alpha=0.58):
    canvas = cv2.cvtColor(gray_u8, cv2.COLOR_GRAY2RGB).astype(np.float32)
    fg = mask_bin > 0
    col = np.array(rgb_color, dtype=np.float32)
    canvas[fg] = (1.0 - alpha) * canvas[fg] + alpha * col
    return np.clip(canvas, 0, 255).astype(np.uint8)

fig, axes = plt.subplots(nrows=len(sample_20), ncols=3, figsize=(13, 3.8 * len(sample_20)))
for idx, rec in enumerate(sample_20):
    axes[idx, 0].imshow(cv2.cvtColor(rec["img"], cv2.COLOR_GRAY2RGB))
    axes[idx, 0].set_title(f"[{idx+1:02d}] {rec['file_name']}", fontsize=9, fontweight="bold")
    axes[idx, 0].axis("off")

    axes[idx, 1].imshow(overlay_mask(rec["img"], rec["gt"], (255, 30, 30)))
    axes[idx, 1].set_title("Ground Truth (Red)", fontsize=9, color="#b30000")
    axes[idx, 1].axis("off")

    axes[idx, 2].imshow(overlay_mask(rec["img"], rec["pred"], (40, 235, 60)))
    axes[idx, 2].set_title(f"EffNet-B3 + DeepSup | Dice: {rec['dice']:.3f}", fontsize=9, color="#006600", fontweight="bold")
    axes[idx, 2].axis("off")

plt.tight_layout()
plt.savefig(TEST_VIS_FIG_PATH, dpi=100, bbox_inches="tight")
print(f"[+] Saved all outputs inside: {SAVE_DIR}/")
```

```python
import os
os.environ["OPENCV_LOG_LEVEL"] = "SILENT"
os.environ["PYTORCH_CUDA_ALLOC_CONF"] = "expandable_segments:True"

import json
import time
import random
from pathlib import Path
from concurrent.futures import ThreadPoolExecutor

import cv2
import numpy as np
import pandas as pd
import matplotlib
matplotlib.use("Agg")
import matplotlib.pyplot as plt
from tqdm import tqdm

import torch
import torch.nn as nn
import torch.nn.functional as F
import segmentation_models_pytorch as smp
from segmentation_models_pytorch.base import SegmentationHead
from segment_anything import sam_model_registry

# =====================================================================
# 1. Paths & Hyperparameters (SEED = 42, 512 x 512)
# =====================================================================
SEED = 42
TARGET_SHAPE = (512, 512)

# Morphology-Aware Box Extraction
MIN_BLOB_AREA = 5
PAD_RATIO = 0.25
PAD_MIN, PAD_MAX = 6, 16

random.seed(SEED)
np.random.seed(SEED)
torch.manual_seed(SEED)
torch.cuda.manual_seed_all(SEED)

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
use_bf16 = torch.cuda.is_available() and torch.cuda.is_bf16_supported()
amp_dtype = torch.bfloat16 if use_bf16 else torch.float16
print(f"[+] Device: {device} ({torch.cuda.get_device_name(0) if torch.cuda.is_available() else 'CPU'}) | AMP: {amp_dtype}")

DATASET_ROOT = Path("./FracAtlas")
TEST_CSV = Path("./splits/test_split_50_50.csv")

EFFNET_DIR = Path("./FracAtlas_UNetPP_EfficientNetB3_DeepSup")
EFFNET_CKPT = EFFNET_DIR / "unetpp_effnetb3_deepsup_best.pt"

MEDSAM_BASE_CKPT = Path("./medsam_vit_b.pth")
MEDSAM_LORA_CKPT = Path("./FracAtlas_MedSAM_LoRA/medsam_lora_r8_binary_best.pt")

OUT_CSV_PATH = EFFNET_DIR / "dual_model_effnetb3_medsam_benchmark.csv"
OUT_FIG_PATH = EFFNET_DIR / "dual_model_effnetb3_medsam_20.png"

for p in [EFFNET_CKPT, MEDSAM_BASE_CKPT, MEDSAM_LORA_CKPT, TEST_CSV]:
    if not p.exists():
        raise FileNotFoundError(f"Required file not found: {p}")

def extract_clean_state_dict(ckpt_path):
    payload = torch.load(ckpt_path, map_location="cpu")
    state = payload["model_state_dict"] if (isinstance(payload, dict) and "model_state_dict" in payload) else payload
    return {k.replace("_orig_mod.", ""): v for k, v in state.items()}

# =====================================================================
# 2. Load Only Stage 1 (U-Net++ EffNet-B3 DeepSup) & Stage 2 (MedSAM-LoRA)
# =====================================================================
class UNetPlusPlusDeepSupervision(nn.Module):
    def __init__(self, encoder_name="efficientnet-b3", encoder_weights=None, classes=3, input_size=512):
        super().__init__()
        self.net = smp.UnetPlusPlus(
            encoder_name=encoder_name,
            encoder_weights=encoder_weights,
            in_channels=3,
            classes=classes,
            decoder_channels=(256, 128, 64, 32, 16),
        )
        top_keys = sorted(
            [k for k in self.net.decoder.blocks.keys() if k.startswith("x_0_") and k != "x_0_0"],
            key=lambda s: int(s.split("_")[-1])
        )
        self.aux_keys = top_keys[:-1]
        self._aux_feats = {}
        self._capture_for_init = True

        for k in self.aux_keys:
            self.net.decoder.blocks[k].register_forward_hook(self._make_hook(k))

        with torch.no_grad():
            _ = self.net(torch.zeros(1, 3, input_size, input_size))

        self.aux_heads = nn.ModuleDict()
        for k in self.aux_keys:
            feat = self._aux_feats[k]
            self.aux_heads[k] = SegmentationHead(
                in_channels=feat.shape[1],
                out_channels=classes,
                kernel_size=3,
                upsampling=input_size // feat.shape[2],
            )
        self._aux_feats.clear()
        self._capture_for_init = False

    def _make_hook(self, key):
        def hook(_module, _inp, output):
            if self.training or self._capture_for_init:
                self._aux_feats[key] = output
        return hook

    def forward(self, x):
        return self.net(x)

eff_model = UNetPlusPlusDeepSupervision("efficientnet-b3", encoder_weights=None, classes=3, input_size=512)
eff_model.load_state_dict(extract_clean_state_dict(EFFNET_CKPT))
eff_model = eff_model.float().to(device).eval()

def resize_medsam_to_512(sam_model, target_size=512):
    new_grid = target_size // 16
    sam_model.image_encoder.img_size = target_size
    sam_model.prompt_encoder.input_image_size = (target_size, target_size)
    sam_model.prompt_encoder.image_embedding_size = (new_grid, new_grid)

    pos_embed = sam_model.image_encoder.pos_embed.permute(0, 3, 1, 2)
    pos_embed_new = F.interpolate(pos_embed, size=(new_grid, new_grid), mode="bicubic", align_corners=False)
    sam_model.image_encoder.pos_embed = nn.Parameter(pos_embed_new.permute(0, 2, 3, 1))

    new_rel_len = 2 * new_grid - 1
    for blk in sam_model.image_encoder.blocks:
        if hasattr(blk.attn, "rel_pos_h") and blk.attn.rel_pos_h is not None and blk.attn.rel_pos_h.shape[0] != new_rel_len:
            rh = blk.attn.rel_pos_h.permute(1, 0).unsqueeze(0)
            blk.attn.rel_pos_h = nn.Parameter(
                F.interpolate(rh, size=new_rel_len, mode="linear", align_corners=False).squeeze(0).permute(1, 0)
            )
            rw = blk.attn.rel_pos_w.permute(1, 0).unsqueeze(0)
            blk.attn.rel_pos_w = nn.Parameter(
                F.interpolate(rw, size=new_rel_len, mode="linear", align_corners=False).squeeze(0).permute(1, 0)
            )

class LoRAQKVLinear(nn.Module):
    def __init__(self, qkv_linear: nn.Linear, r: int = 8, alpha: int = 16):
        super().__init__()
        self.qkv = qkv_linear
        self.scaling = alpha / float(r)
        self.lora_A = nn.Parameter(torch.zeros((r, qkv_linear.in_features)))
        self.lora_B = nn.Parameter(torch.zeros((qkv_linear.out_features, r)))

    def forward(self, x):
        return self.qkv(x) + F.linear(F.linear(x, self.lora_A), self.lora_B) * self.scaling

class MedSAMLoRABinary(nn.Module):
    def __init__(self, checkpoint_path, rank=8, alpha=16):
        super().__init__()
        self.sam = sam_model_registry["vit_b"](checkpoint=str(checkpoint_path))
        resize_medsam_to_512(self.sam, target_size=TARGET_SHAPE[0])
        for blk in self.sam.image_encoder.blocks:
            blk.attn.qkv = LoRAQKVLinear(blk.attn.qkv, r=rank, alpha=alpha)

    @torch.inference_mode()
    def decode_boxes_prob_map(self, image_embeddings, boxes_list):
        if len(boxes_list) == 0:
            return np.zeros(TARGET_SHAPE, dtype=np.float32)

        K = len(boxes_list)
        boxes_t = torch.tensor(boxes_list, dtype=torch.float32, device=device).unsqueeze(1)
        sparse_emb, dense_emb = self.sam.prompt_encoder(points=None, boxes=boxes_t, masks=None)

        dec = self.sam.mask_decoder
        dense_pe = self.sam.prompt_encoder.get_dense_pe()
        output_tokens = torch.cat([dec.iou_token.weight, dec.mask_tokens.weight], dim=0).unsqueeze(0).expand(K, -1, -1)
        tokens = torch.cat((output_tokens, sparse_emb), dim=1)

        src = image_embeddings.expand(K, -1, -1, -1) + dense_emb
        pos_src = dense_pe.expand(K, -1, -1, -1)
        b, c, h, w = src.shape

        hs, src = dec.transformer(src, pos_src, tokens)
        mask_tokens_out = hs[:, 1:2, :]
        src = src.transpose(1, 2).view(b, c, h, w)
        upscaled = dec.output_upscaling(src)
        hyper_in = dec.output_hypernetworks_mlps[0](mask_tokens_out[:, 0, :]).unsqueeze(1)

        b, c_up, h_up, w_up = upscaled.shape
        low_res = (hyper_in @ upscaled.view(b, c_up, h_up * w_up)).view(b, 1, h_up, w_up)
        up_logits = F.interpolate(low_res.float(), size=TARGET_SHAPE, mode="bilinear", align_corners=False)
        return torch.sigmoid(up_logits).squeeze(1).max(dim=0).values.cpu().numpy().astype(np.float32)

medsam_lora = MedSAMLoRABinary(MEDSAM_BASE_CKPT, rank=8, alpha=16)
medsam_lora.load_state_dict(extract_clean_state_dict(MEDSAM_LORA_CKPT))
medsam_lora = medsam_lora.to(device).eval()
print("[+] Loaded 2-Model Pipeline: U-Net++ (EffNet-B3 + DeepSup) + MedSAM-LoRA (r=8) [No ResNet, No YOLO]")

# =====================================================================
# 3. Load & Cache Hold-Out Test Split (CLAHE + 512x512, N = 216)
# =====================================================================
json_candidates = list((DATASET_ROOT / "Annotations").rglob("*COCO*.json")) or list((DATASET_ROOT / "Annotations").rglob("*.json"))
with open(json_candidates[0], "r") as f:
    coco_payload = json.load(f)

img_id_to_annos = {}
for ann in coco_payload["annotations"]:
    img_id_to_annos.setdefault(ann["image_id"], []).append(ann)

test_df = pd.read_csv(TEST_CSV).reset_index(drop=True)

def _load_item(row_dict):
    clahe = cv2.createCLAHE(clipLimit=2.0, tileGridSize=(8, 8))
    gray = cv2.imread(str(row_dict["path"]), cv2.IMREAD_GRAYSCALE)
    orig_h, orig_w = gray.shape
    clahe_512 = cv2.resize(clahe.apply(gray), TARGET_SHAPE, interpolation=cv2.INTER_CUBIC)
    norm = (clahe_512.astype(np.float32) - clahe_512.min()) / (clahe_512.max() - clahe_512.min() + 1e-8)

    gt_nat = np.zeros((orig_h, orig_w), dtype=np.uint8)
    coco_id = int(row_dict["coco_id"])
    if coco_id != -1 and coco_id in img_id_to_annos:
        for ann in img_id_to_annos[coco_id]:
            for poly in ann.get("segmentation", []):
                pts = np.array(poly, dtype=np.int32).reshape((-1, 2))
                cv2.fillPoly(gt_nat, [pts], 1)

    gt_512 = cv2.resize(gt_nat, TARGET_SHAPE, interpolation=cv2.INTER_NEAREST)
    return {
        "file_name": row_dict["file_name"],
        "fractured": int(row_dict["fractured"]),
        "resized_u8": clahe_512,
        "norm_f32": norm,
        "gt_512": gt_512,
    }

with ThreadPoolExecutor(max_workers=min(8, os.cpu_count() or 4)) as ex:
    cached_test = list(ex.map(_load_item, test_df.to_dict(orient="records")))

# =====================================================================
# 4. 2-View TTA, Morphology-Aware Boxes & Metrics
# =====================================================================
@torch.inference_mode()
def predict_2view_tta(model, inp_512):
    inp_flip = torch.flip(inp_512, dims=[3])
    probs = torch.softmax(model(torch.cat([inp_512, inp_flip], dim=0)), dim=1)
    p1 = probs[0:1]
    p2 = torch.flip(probs[1:2], dims=[3])
    p_avg = 0.5 * (p1 + p2)
    p_frac = p_avg[0, 2].cpu().numpy().astype(np.float32)
    mask_bin = (torch.argmax(p_avg, dim=1).squeeze(0) == 2).cpu().numpy().astype(np.uint8)
    return p_frac, mask_bin

def extract_effnet_boxes_and_roi(mask_bin):
    H, W = TARGET_SHAPE
    num_labels, _, stats, _ = cv2.connectedComponentsWithStats(mask_bin, connectivity=8)
    boxes = []
    roi = np.zeros(TARGET_SHAPE, dtype=np.float32)
    for lbl in range(1, num_labels):
        if stats[lbl, cv2.CC_STAT_AREA] >= MIN_BLOB_AREA:
            x, y, w, h = int(stats[lbl, 0]), int(stats[lbl, 1]), int(stats[lbl, 2]), int(stats[lbl, 3])
            px = int(np.clip(round(PAD_RATIO * w), PAD_MIN, PAD_MAX))
            py = int(np.clip(round(PAD_RATIO * h), PAD_MIN, PAD_MAX))
            x1, y1 = max(0, x - px), max(0, y - py)
            x2, y2 = min(W - 1, x + w + px), min(H - 1, y + h + py)
            boxes.append([x1, y1, x2, y2])
            roi[y1 : y2 + 1, x1 : x2 + 1] = 1.0
    return boxes, roi

DIAG_PENALTY = float(np.hypot(TARGET_SHAPE[0], TARGET_SHAPE[1]))
BOUNDARY_KERNEL = np.ones((3, 3), dtype=np.uint8)

def compute_metrics(pred_bin, gt_bin, smooth=1e-5):
    p = (pred_bin > 0).astype(np.uint8)
    g = (gt_bin > 0).astype(np.uint8)
    inter = int(np.logical_and(p, g).sum())
    p_sum, g_sum = int(p.sum()), int(g.sum())
    union = int(np.logical_or(p, g).sum())

    dice = (2.0 * inter + smooth) / (p_sum + g_sum + smooth)
    iou = (inter + smooth) / (union + smooth)

    if p_sum == 0 and g_sum == 0:
        hd95 = 0.0
    elif p_sum == 0 or g_sum == 0:
        hd95 = DIAG_PENALTY
    else:
        p_bnd = p - cv2.erode(p, BOUNDARY_KERNEL, iterations=1)
        g_bnd = g - cv2.erode(g, BOUNDARY_KERNEL, iterations=1)
        if p_bnd.sum() == 0:
            p_bnd = p
        if g_bnd.sum() == 0:
            g_bnd = g
        dt_g = cv2.distanceTransform(1 - g_bnd, cv2.DIST_L2, 5)
        dt_p = cv2.distanceTransform(1 - p_bnd, cv2.DIST_L2, 5)
        dists = np.concatenate([dt_g[p_bnd > 0], dt_p[g_bnd > 0]])
        hd95 = float(np.percentile(dists, 95))

    return float(dice), float(iou), hd95, inter, p_sum, g_sum

# =====================================================================
# 5. Run Dual-Model Inference & Cache Probability Maps (N = 216)
# =====================================================================
cached_data = []
latencies_ms = []

with torch.inference_mode():
    for item in tqdm(cached_test, desc="Dual-Model Inference (EffNet-B3 -> MedSAM-LoRA)"):
        t0 = time.time()
        inp = torch.from_numpy(item["norm_f32"]).unsqueeze(0).unsqueeze(0).repeat(1, 3, 1, 1).to(device)

        p_eff, m_eff = predict_2view_tta(eff_model, inp)
        boxes, roi = extract_effnet_boxes_and_roi(m_eff)

        if len(boxes) > 0:
            with torch.amp.autocast("cuda", dtype=amp_dtype, enabled=(device.type == "cuda")):
                emb = medsam_lora.sam.image_encoder(inp.to(dtype=amp_dtype))
                p_sam = medsam_lora.decode_boxes_prob_map(emb, boxes)
        else:
            p_sam = np.zeros(TARGET_SHAPE, dtype=np.float32)

        if device.type == "cuda":
            torch.cuda.synchronize()
        latencies_ms.append((time.time() - t0) * 1000.0)

        cached_data.append({
            "p_eff": p_eff,
            "m_eff": m_eff,
            "p_sam": p_sam,
            "boxes": boxes,
            "roi": roi,
        })

print(f"[+] Mean End-to-End Dual-Model Latency: {np.mean(latencies_ms):.1f} ms/scan")

# =====================================================================
# 6. Sweep Residual Beta & Thresholds 
# =====================================================================
rows = []
vis_store = {}

def eval_rule(name, beta, thresh, pred_fn):
    dices, ious, hd95s, healthy_fps = [], [], [], []
    m_inter = m_pred = m_gt = 0
    det_scans = 0
    preds = []

    for item, cd in zip(cached_test, cached_data):
        pred_bin = pred_fn(cd)
        preds.append(pred_bin)

        if item["fractured"] == 1:
            if pred_bin.sum() > 0:
                det_scans += 1
            d, iou, h95, inter, p_s, g_s = compute_metrics(pred_bin, item["gt_512"])
            dices.append(d); ious.append(iou); hd95s.append(h95)
            m_inter += inter; m_pred += p_s; m_gt += g_s
        else:
            healthy_fps.append(float(pred_bin.sum()))

    micro_dice = (2.0 * m_inter + 1e-5) / (m_pred + m_gt + 1e-5)
    rows.append({
        "Pipeline Configuration": name,
        "MedSAM Beta": beta,
        "Threshold": thresh,
        "Scan Recall": f"{det_scans}/108 ({det_scans/108*100:.1f}%)",
        "Macro-Dice": round(float(np.mean(dices)), 4),
        "Micro-Dice": round(float(micro_dice), 4),
        "Macro-IoU": round(float(np.mean(ious)), 4),
        "HD95 (px)": round(float(np.mean(hd95s)), 2),
        "Healthy FP (px)": round(float(np.mean(healthy_fps)), 2),
    })
    vis_store[(name, beta, thresh)] = preds

# Baseline: Standalone EffNet-B3 (No MedSAM)
eval_rule("Standalone U-Net++ EffNet-B3 (2V-TTA)", 0.00, 0.50, lambda cd: cd["m_eff"])

# Dual-Model Hybrid: P_final = P_EffNet + beta * ROI * (P_MedSAM - 0.50)
for beta in [0.10, 0.15, 0.20, 0.25, 0.30]:
    for thresh in [0.45, 0.50]:
        eval_rule(
            "Dual-Model: P_EffNet + beta*ROI*(P_MedSAM - 0.50)",
            beta,
            thresh,
            lambda cd, b=beta, t=thresh: ((cd["p_eff"] + b * cd["roi"] * (cd["p_sam"] - 0.50)) > t).astype(np.uint8),
        )

df_res = pd.DataFrame(rows).sort_values(by="Macro-Dice", ascending=False).reset_index(drop=True)
print("\n" + "=" * 136)
print("DUAL-MODEL PIPELINE BENCHMARK: U-Net++ (EffNet-B3) + MedSAM-LoRA ONLY (No ResNet-34, No YOLO | N = 216)")
print("=" * 136)
print(df_res.to_string(index=False))
print("=" * 136)

df_res.to_csv(OUT_CSV_PATH, index=False)
best_row = df_res.iloc[0]
print(f"\n[+] PEAK DUAL-MODEL CONFIG: {best_row['Pipeline Configuration']} (beta={best_row['MedSAM Beta']}, t={best_row['Threshold']})")
print(f"    -> Recall: {best_row['Scan Recall']} | Macro-Dice: {best_row['Macro-Dice']:.4f} | Micro-Dice: {best_row['Micro-Dice']:.4f} | IoU: {best_row['Macro-IoU']:.4f} | HD95: {best_row['HD95 (px)']} px | Healthy FP: {best_row['Healthy FP (px)']} px")
print(f"[+] Saved benchmark CSV to: {OUT_CSV_PATH}")

# =====================================================================
# 7. Save 20-Scan Qualitative Visual Comparison
# =====================================================================
frac_indices = [i for i, r in enumerate(cached_test) if r["fractured"] == 1]
rng = random.Random(SEED)
sample_20 = rng.sample(frac_indices, k=min(20, len(frac_indices)))

def overlay_mask(gray_u8, mask_bin, rgb_color, alpha=0.58):
    canvas = cv2.cvtColor(gray_u8, cv2.COLOR_GRAY2RGB).astype(np.float32)
    fg = mask_bin > 0
    col = np.array(rgb_color, dtype=np.float32)
    canvas[fg] = (1.0 - alpha) * canvas[fg] + alpha * col
    return np.clip(canvas, 0, 255).astype(np.uint8)

best_key = (best_row["Pipeline Configuration"], float(best_row["MedSAM Beta"]), float(best_row["Threshold"]))

fig, axes = plt.subplots(nrows=len(sample_20), ncols=3, figsize=(13, 3.8 * len(sample_20)))
for row_i, g_idx in enumerate(sample_20):
    item = cached_test[g_idx]
    cd = cached_data[g_idx]
    gt = item["gt_512"]

    m_eff = vis_store[("Standalone U-Net++ EffNet-B3 (2V-TTA)", 0.00, 0.50)][g_idx]
    m_best = vis_store[best_key][g_idx]

    d_eff, _, _, _, _, _ = compute_metrics(m_eff, gt)
    d_best, _, _, _, _, _ = compute_metrics(m_best, gt)

    gt_vis = overlay_mask(item["resized_u8"], gt, (255, 30, 30))
    eff_vis = overlay_mask(item["resized_u8"], m_eff, (40, 170, 255))
    best_vis = overlay_mask(item["resized_u8"], m_best, (40, 240, 40))

    for (x1, y1, x2, y2) in cd["boxes"]:
        cv2.rectangle(best_vis, (x1, y1), (x2, y2), (255, 225, 0), 2)

    axes[row_i, 0].imshow(gt_vis)
    axes[row_i, 0].set_title(f"[{row_i+1:02d}] GT | {item['file_name']}", fontsize=9, color="#b30000", fontweight="bold")
    axes[row_i, 0].axis("off")

    axes[row_i, 1].imshow(eff_vis)
    axes[row_i, 1].set_title(f"EffNet-B3 Only | Dice: {d_eff:.3f}", fontsize=9, color="#005599")
    axes[row_i, 1].axis("off")

    axes[row_i, 2].imshow(best_vis)
    axes[row_i, 2].set_title(f"EffNet-B3 + MedSAM-LoRA | Dice: {d_best:.3f}", fontsize=9, color="#006600", fontweight="bold")
    axes[row_i, 2].axis("off")

plt.tight_layout()
plt.savefig(OUT_FIG_PATH, dpi=100, bbox_inches="tight")
print(f"[+] Saved 20-scan dual-model comparison grid to: {OUT_FIG_PATH}")
```
